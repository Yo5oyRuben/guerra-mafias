
# Apendice: analisis del retrato de fases de la ecuacion replicadora de dos variables

Esta notebook desarrolla la parte del apendice mostrada en la imagen. La idea no es transcribir formulas sueltas, sino reconstruir el argumento completo:

1. que representa el sistema dinamico;
2. por que las nulclinas organizan el retrato de fases;
3. como se obtienen los puntos fijos como intersecciones de nulclinas;
4. que cruces se conservan y cuales se descartan;
5. para que sirve el jacobiano en el analisis local de estabilidad;
6. que significa la condicion degenerada de coincidencia de nulclinas.

Mantendremos la notacion del apendice:

- $x_1,x_2$ son las variables dinamicas, con $0 \le x_1,x_2 \le 1$.
- $b,r,\epsilon,\beta,p$ son parametros.
- $F_1,F_2$ son las velocidades reescaladas.
- Los puntos fijos se denotan $A,B,C,D,A',B',E$.

Para aligerar las derivaciones usaremos tambien, solo como abreviaturas internas,

$$
M=b-1-r,\qquad N=b-1-\epsilon,
$$

porque en el apendice aparecen repetidamente $b-1-r$ y $b-1-\epsilon$.


```python

import sympy as sp
from IPython.display import display

sp.init_printing()

x1, x2 = sp.symbols("x_1 x_2", real=True)
b, r, eps, beta, p = sp.symbols("b r epsilon beta p", real=True)

M = b - 1 - r
N = b - 1 - eps
S = beta*r + p*eps
T = r + beta*p*eps


def simp(expr):
    '''Simplificacion corta para las comprobaciones algebraicas.'''
    return sp.factor(sp.simplify(expr))


def check_zero(label, expr):
    value = simp(expr)
    print(f"{label}: {value}")
```


## 1. Sistema dinamico: ecuaciones (A1) y (A2)

El apendice empieza escribiendo la ecuacion replicadora reescalada como un campo de velocidades bidimensional:

$$
\dot{x}_1=F_1(x_1,x_2),\qquad \dot{x}_2=F_2(x_1,x_2). \tag{A1}
$$

Las velocidades son

$$
F_1(x_1,x_2)=x_1(1-x_1)\left\{\beta[x_1(1-b+r)-r]+p[x_2(1-b+\epsilon)-\epsilon]\right\},
$$

$$
F_2(x_1,x_2)=x_2(1-x_2)\left\{x_2(1-b+r)-r+\beta p[x_1(1-b+\epsilon)-\epsilon]\right\}. \tag{A2}
$$

Conceptualmente, $F_i$ dice hacia donde se mueve la poblacion $i$ en el eje $x_i$. El factor $x_i(1-x_i)$ es el factor replicador de dos estrategias: si $x_i=0$ o $x_i=1$, la poblacion ya esta en un estado puro y no puede salirse del intervalo. Por eso los bordes del cuadrado unidad son invariantes.

La parte entre llaves es el gradiente efectivo de seleccion. Si es positivo, $x_i$ aumenta; si es negativo, $x_i$ disminuye; si es cero, la velocidad de esa coordenada se anula aunque $x_i$ sea interior.


```python

g1 = beta*(x1*(1 - b + r) - r) + p*(x2*(1 - b + eps) - eps)
g2 = x2*(1 - b + r) - r + beta*p*(x1*(1 - b + eps) - eps)

F1 = x1*(1 - x1)*g1
F2 = x2*(1 - x2)*g2

display(sp.Eq(sp.Symbol("F_1"), sp.factor(F1)))
display(sp.Eq(sp.Symbol("F_2"), sp.factor(F2)))

# Las mismas expresiones usando M=b-1-r, N=b-1-epsilon,
# S=beta*r+p*epsilon y T=r+beta*p*epsilon.
check_zero("g1 - (-beta*M*x1 - p*N*x2 - S)", g1 - (-beta*M*x1 - p*N*x2 - S))
check_zero("g2 - (-beta*p*N*x1 - M*x2 - T)", g2 - (-beta*p*N*x1 - M*x2 - T))
```


## 2. Nulclinas: que son y por que se imponen $F_i=0$

Una nulclina de $F_i$ es el conjunto de puntos donde la componente $i$ de la velocidad se anula:

$$
F_i(x_1,x_2)=0.
$$

En el retrato de fases esto es util porque sobre la nulclina de $F_1$ el campo no tiene componente horizontal, y sobre la nulclina de $F_2$ no tiene componente vertical. Un punto fijo exige ambas cosas a la vez:

$$
F_1(x_1,x_2)=0,\qquad F_2(x_1,x_2)=0.
$$

Por la factorizacion de (A2), cada nulclina sale de anular uno de los factores. Para $F_1$ tenemos

$$
F_1=x_1(1-x_1)g_1(x_1,x_2),
$$

con

$$
g_1=\beta[x_1(1-b+r)-r]+p[x_2(1-b+\epsilon)-\epsilon].
$$

Entonces $F_1=0$ se cumple si

$$
x_1=0, \tag{A3}
$$

$$
x_1=1, \tag{A4}
$$

o si $g_1=0$. Usando $1-b+r=-(b-1-r)=-M$ y $1-b+\epsilon=-(b-1-\epsilon)=-N$,

$$
g_1=-\beta Mx_1-pNx_2-(\beta r+p\epsilon).
$$

Imponiendo $g_1=0$:

$$
-\beta Mx_1-pNx_2-(\beta r+p\epsilon)=0,
$$

$$
pNx_2=-\beta Mx_1-(\beta r+p\epsilon),
$$

$$
x_2=\frac{-x_1\beta(b-1-r)-(\beta r+p\epsilon)}{p(b-1-\epsilon)}. \tag{A5}
$$

Para $F_2$ ocurre lo mismo:

$$
F_2=x_2(1-x_2)g_2(x_1,x_2),
$$

con

$$
g_2=x_2(1-b+r)-r+\beta p[x_1(1-b+\epsilon)-\epsilon].
$$

Por tanto aparecen las nulclinas de borde

$$
x_2=0, \tag{A6}
$$

$$
x_2=1, \tag{A7}
$$

y, desde $g_2=0$,

$$
-Mx_2-\beta pNx_1-(r+\beta p\epsilon)=0,
$$

$$
x_2=\frac{-x_1\beta p(b-1-\epsilon)-(r+\beta p\epsilon)}{b-1-r}. \tag{A8}
$$


```python

A5 = (-x1*beta*M - S)/(p*N)
A8 = (-x1*beta*p*N - T)/M

print("Nulclinas interiores resueltas para x_2:")
display(sp.Eq(x2, A5))
display(sp.Eq(x2, A8))

print("\nComprobaciones SymPy:")
check_zero("F1 en x1=0", F1.subs(x1, 0))
check_zero("F1 en x1=1", F1.subs(x1, 1))
check_zero("g1 al sustituir (A5)", g1.subs(x2, A5))
check_zero("F2 en x2=0", F2.subs(x2, 0))
check_zero("F2 en x2=1", F2.subs(x2, 1))
check_zero("g2 al sustituir (A8)", g2.subs(x2, A8))

print("\nSolucion directa con solve:")
display(sp.solve(sp.Eq(g1, 0), x2)[0])
display(sp.solve(sp.Eq(g2, 0), x2)[0])
```


## 3. Puntos fijos como intersecciones de nulclinas

Un punto fijo es un punto donde el vector velocidad completo se anula. Como hay dos componentes, necesitamos estar simultaneamente en una nulclina de $F_1$ y en una nulclina de $F_2$.

El apendice organiza las nulclinas en dos grupos:

- grupo $F_1=0$: (A3), (A4), (A5);
- grupo $F_2=0$: (A6), (A7), (A8).

Al cruzar una nulclina del primer grupo con una del segundo se obtienen nueve candidatos algebraicos. Las cuatro esquinas aparecen siempre porque en ellas se anulan los factores replicadores de borde:

$$
A=(0,1),\qquad B=(1,0),\qquad C=(1,1),\qquad D=(0,0).
$$

Estos cuatro puntos no dependen de los parametros. En cambio, los otros candidatos pueden estar dentro o fuera del cuadrado unidad segun los valores de $b,r,\epsilon,\beta,p$.


```python

A = (sp.Integer(0), sp.Integer(1))
B = (sp.Integer(1), sp.Integer(0))
C = (sp.Integer(1), sp.Integer(1))
D = (sp.Integer(0), sp.Integer(0))

A9 = -T/M
A10 = -S/(beta*M)

den_E = (p*N)**2 - M**2
x1_E = (M*S - p*N*T)/(beta*den_E)
x2_E = (M*T - p*N*S)/den_E

Aprime = (sp.Integer(0), A9)
Bprime = (A10, sp.Integer(0))
E = (x1_E, x2_E)

points = {
    "A": A,
    "B": B,
    "C": C,
    "D": D,
    "A'": Aprime,
    "B'": Bprime,
    "E": E,
}

for name, point in points.items():
    print(f"{name} =")
    display(point)

print("\nComprobacion F1=F2=0 en los siete candidatos algebraicos:")
for name, (px, py) in points.items():
    print(f"{name}:")
    check_zero("  F1", F1.subs({x1: px, x2: py}))
    check_zero("  F2", F2.subs({x1: px, x2: py}))
```


## 4. Derivacion de $A'$, $B'$ y $E$

El punto $A'$ es el cruce de (A3) y (A8). Por tanto $x_1(A')=0$ y sustituimos $x_1=0$ en (A8):

$$
x_2(A')=\frac{-(r+\beta p\epsilon)}{b-1-r}. \tag{A9}
$$

El punto $B'$ es el cruce de (A5) y (A6). Por tanto $x_2(B')=0$ y se impone (A5) igual a cero:

$$
0=\frac{-x_1\beta(b-1-r)-(\beta r+p\epsilon)}{p(b-1-\epsilon)}.
$$

El denominador no importa para anular la fraccion en el caso generico; debe anularse el numerador:

$$
-x_1\beta(b-1-r)-(\beta r+p\epsilon)=0,
$$

$$
x_1(B')=-\frac{\beta r+p\epsilon}{\beta(b-1-r)}. \tag{A10}
$$

El punto $E$ es el cruce de las dos nulclinas interiores (A5) y (A8). Conviene escribirlas como sistema lineal. A partir de $g_1=0$ y $g_2=0$:

$$
\beta Mx_1+pNx_2=-(\beta r+p\epsilon),
$$

$$
\beta pNx_1+Mx_2=-(r+\beta p\epsilon).
$$

Con $S=\beta r+p\epsilon$ y $T=r+\beta p\epsilon$, el sistema es

$$
\begin{pmatrix}
\beta M & pN\\
\beta pN & M
\end{pmatrix}
\begin{pmatrix}x_1\\x_2\end{pmatrix}
=
\begin{pmatrix}-S\\-T\end{pmatrix}.
$$

El determinante es

$$
\Delta=\beta\left[M^2-(pN)^2\right].
$$

Aplicando la regla de Cramer y cambiando el signo de numerador y denominador para escribirlo como en el apendice:

$$
x_1(E)=\frac{M S-pN T}{\beta[(pN)^2-M^2]},
$$

$$
x_2(E)=\frac{M T-pN S}{(pN)^2-M^2}.
$$

En la notacion original:

$$
x_1(E)=\frac{(b-1-r)(\beta r+p\epsilon)-p(b-1-\epsilon)(r+\beta p\epsilon)}{\beta\{[p(b-1-\epsilon)]^2-(b-1-r)^2\}}, \tag{A11}
$$

$$
x_2(E)=\frac{(b-1-r)(r+\beta p\epsilon)-p(b-1-\epsilon)(\beta r+p\epsilon)}{[p(b-1-\epsilon)]^2-(b-1-r)^2}. \tag{A12}
$$

Estas expresiones solo representan un punto interior fisico si $0\le x_1(E),x_2(E)\le 1$ y si el denominador no se anula.


```python

A_mat = sp.Matrix([[beta*M, p*N], [beta*p*N, M]])
rhs = sp.Matrix([-S, -T])
sol_cramer = sp.simplify(A_mat.LUsolve(rhs))

display(sp.Eq(sp.Symbol("det"), sp.factor(A_mat.det())))
display(sol_cramer)

print("Comprobacion de las formulas (A11)-(A12) frente a la solucion lineal:")
check_zero("x1 por Cramer - x1(E)", sol_cramer[0] - x1_E)
check_zero("x2 por Cramer - x2(E)", sol_cramer[1] - x2_E)

print("\nComprobacion directa de que E esta en las dos nulclinas interiores:")
check_zero("g1(E)", g1.subs({x1: x1_E, x2: x2_E}))
check_zero("g2(E)", g2.subs({x1: x1_E, x2: x2_E}))
```


## 5. Cruces descartados y papel del cuadrado unidad

Los nueve cruces algebraicos no son automaticamente relevantes. El modelo solo interpreta soluciones dentro del cuadrado unidad:

$$
0\le x_1\le 1,\qquad 0\le x_2\le 1.
$$

El apendice conserva $A,B,C,D,A',B',E$ y descarta dos cruces:

1. (A4) con (A8): fija $x_1=1$ y corta la nulclina interior de $F_2$.
2. (A5) con (A7): fija $x_2=1$ y corta la nulclina interior de $F_1$.

Veamos por que quedan fuera para el rango usual del apendice, con $b>1$, $r\ge 0$, $\beta>0$ y $p\ge 0$.

Para (A4)-(A8):

$$
x_2=\frac{-\beta p(b-1-\epsilon)-(r+\beta p\epsilon)}{b-1-r}
=-\frac{\beta p(b-1)+r}{b-1-r}.
$$

Si $b-1-r>0$, entonces $x_2<0$. Si $b-1-r<0$, entonces

$$
x_2-1=-\frac{(1+\beta p)(b-1)}{b-1-r}>0,
$$

por lo que $x_2>1$. En ambos casos esta fuera de $[0,1]$.

Para (A5)-(A7):

$$
x_1=-\frac{\beta r+p(b-1)}{\beta(b-1-r)}.
$$

Si $b-1-r>0$, entonces $x_1<0$. Si $b-1-r<0$, entonces

$$
x_1-1=-\frac{(\beta+p)(b-1)}{\beta(b-1-r)}>0,
$$

por lo que $x_1>1$. Tambien queda fuera del cuadrado unidad.


```python

x2_A4_A8 = simp(A8.subs(x1, 1))
x1_A5_A7 = simp(sp.solve(sp.Eq(A5, 1), x1)[0])

print("Cruce (A4)-(A8):")
display(sp.Eq(x2, x2_A4_A8))
check_zero("forma simplificada", x2_A4_A8 + (beta*p*(b - 1) + r)/M)
display(sp.Eq(x2 - 1, simp(x2_A4_A8 - 1)))

print("\nCruce (A5)-(A7):")
display(sp.Eq(x1, x1_A5_A7))
check_zero("forma simplificada", x1_A5_A7 + (beta*r + p*(b - 1))/(beta*M))
display(sp.Eq(x1 - 1, simp(x1_A5_A7 - 1)))
```


## 6. Jacobiano: ecuacion (A13) y estabilidad local

Una vez localizados los puntos fijos, el apendice estudia su estabilidad con perturbaciones pequenas. Si $x^*=(x_1^*,x_2^*)$ es un punto fijo y escribimos

$$
x=x^*+\delta x,
$$

entonces una expansion de Taylor del campo de velocidades da

$$
\dot{\delta x}=J(x^*)\delta x+O(\|\delta x\|^2).
$$

La matriz lineal $J(x^*)$ es el jacobiano:

$$
J(x^*)=
\left.
\begin{pmatrix}
\dfrac{\partial F_1}{\partial x_1} & \dfrac{\partial F_1}{\partial x_2}\\[6pt]
\dfrac{\partial F_2}{\partial x_1} & \dfrac{\partial F_2}{\partial x_2}
\end{pmatrix}
\right|_{x=x^*}. \tag{A13}
$$

La razon de usar $J$ es que, cerca de un punto fijo hiperbolico, la dinamica no lineal tiene el mismo tipo cualitativo que el sistema lineal. Sus autovalores indican si las perturbaciones crecen o decrecen.

En dimension dos se puede usar la traza y el determinante:

$$
\tau=\operatorname{tr}J,\qquad \Delta_J=\det J.
$$

El polinomio caracteristico es

$$
\lambda^2-\tau\lambda+\Delta_J=0.
$$

Como regla local:

- si $\Delta_J<0$, el punto fijo es una silla;
- si $\Delta_J>0$ y $\tau<0$, las perturbaciones decaen y el punto es estable;
- si $\Delta_J>0$ y $\tau>0$, las perturbaciones crecen y el punto es inestable;
- si algun autovalor tiene parte real cero, aparece un caso no hiperbolico y puede haber bifurcacion o estabilidad marginal.


```python

J = sp.Matrix([
    [sp.diff(F1, x1), sp.diff(F1, x2)],
    [sp.diff(F2, x1), sp.diff(F2, x2)],
])

J_manual = sp.Matrix([
    [(1 - 2*x1)*g1 + x1*(1 - x1)*sp.diff(g1, x1), x1*(1 - x1)*sp.diff(g1, x2)],
    [x2*(1 - x2)*sp.diff(g2, x1), (1 - 2*x2)*g2 + x2*(1 - x2)*sp.diff(g2, x2)],
])

print("Jacobiano en forma factorizada:")
display(sp.simplify(J_manual))

print("\nComprobacion de que coincide con la diferenciacion directa:")
for i in range(2):
    for j in range(2):
        check_zero(f"J[{i},{j}] directo - manual", J[i, j] - J_manual[i, j])
```


## 7. Ejemplo simbolico: jacobiano en las cuatro esquinas

En las esquinas los terminos fuera de la diagonal se anulan porque contienen $x_i(1-x_i)$. Por eso el jacobiano queda diagonal y sus autovalores son simplemente las entradas diagonales.

Esto ilustra como el apendice conecta puntos fijos con estabilidad: primero se enumeran los equilibrios; despues se evalua $J$ en cada uno; finalmente se estudian signos de autovalores, traza y determinante para localizar cambios de estabilidad y bifurcaciones.


```python

for name in ["A", "B", "C", "D"]:
    px, py = points[name]
    Jp = sp.simplify(J_manual.subs({x1: px, x2: py}))
    tau = simp(Jp.trace())
    detJ = simp(Jp.det())
    print(f"{name} = {points[name]}")
    display(Jp)
    display(sp.Eq(sp.Symbol("traza"), tau))
    display(sp.Eq(sp.Symbol("determinante"), detJ))
    print()
```


## 8. Condicion degenerada: coincidencia de nulclinas interiores

El apendice menciona un caso no generico: las dos nulclinas interiores (A5) y (A8) pueden coincidir. Esto no es un simple cruce: si las rectas coinciden, no hay un unico punto $E$, sino todo un segmento de puntos que satisfacen simultaneamente

$$
g_1(x_1,x_2)=0,\qquad g_2(x_1,x_2)=0.
$$

Dentro del cuadrado unidad, ese segmento es un continuo de puntos fijos. Como el campo se anula a lo largo de una direccion tangente al segmento, la linealizacion tiene un autovalor cero en esa direccion. Esa es la razon de hablar de equilibrios marginalmente estables.

Con las abreviaturas $M=b-1-r$, $N=b-1-\epsilon$, $S=\beta r+p\epsilon$ y $T=r+\beta p\epsilon$, las nulclinas interiores son las dos rectas

$$
\beta Mx_1+pNx_2=-S,
$$

$$
\beta pNx_1+Mx_2=-T.
$$

Para que coincidan, primero sus pendientes deben ser iguales. Eso exige

$$
M^2=(pN)^2.
$$

En la rama usada en el apendice,

$$
M=pN,
$$

lo que implica

$$
b-1-r=p(b-1-\epsilon),
$$

$$
(1-p)(b-1)=r-p\epsilon,
$$

$$
b^c(\epsilon,r;\beta=1,p)=1+\frac{r-p\epsilon}{1-p}.
$$

Ademas, para que no sean solo paralelas sino exactamente la misma recta, sus terminos independientes tambien deben coincidir. En esta rama eso exige

$$
S=T.
$$

Como

$$
S-T=(\beta r+p\epsilon)-(r+\beta p\epsilon)=(\beta-1)(r-p\epsilon),
$$

la coincidencia generica se obtiene con

$$
\beta=1.
$$

El apendice anade la condicion $r>-p\epsilon$ para seleccionar el regimen fisico en el que esta coincidencia da lugar al escenario no generico descrito alli. Algebraicamente, el mecanismo es el mismo: bajo $\beta=1$ y en $b=b^c$ (con $p\ne 1$), las dos ecuaciones interiores se reducen a la misma recta. Por eso el punto $E$ deja de estar definido como interseccion aislada y aparece un segmento de equilibrios.


```python

bc = 1 + (r - p*eps)/(1 - p)

print("Comprobaciones de la coincidencia en beta=1 y b=b^c:")
check_zero("M - p*N", (M - p*N).subs(b, bc))
check_zero("S - T con beta=1", (S - T).subs(beta, 1))
check_zero("denominador de E", den_E.subs(b, bc))

# Si beta=1 y b=b^c, las partes interiores g1 y g2 son la misma recta.
g1_degen = g1.subs({beta: 1, b: bc})
g2_degen = g2.subs({beta: 1, b: bc})
check_zero("g1 - g2 en el caso degenerado", g1_degen - g2_degen)

display(sp.Eq(sp.Symbol("g_1"), simp(g1_degen)))
display(sp.Eq(sp.Symbol("g_2"), simp(g2_degen)))
```


## 9. Resumen de la logica del apendice

El argumento completo queda asi:

1. La ecuacion replicadora (A1)-(A2) define un campo de velocidades en el cuadrado unidad.
2. Los factores $x_i(1-x_i)$ hacen invariantes los bordes y producen nulclinas de borde.
3. Las partes interiores $g_1=0$ y $g_2=0$ producen las rectas (A5) y (A8).
4. Los puntos fijos son intersecciones entre una nulclina de $F_1$ y una nulclina de $F_2$.
5. Las cuatro esquinas $A,B,C,D$ siempre existen; $A'$, $B'$ y $E$ dependen de los parametros.
6. Dos cruces algebraicos se descartan porque quedan fuera de $0\le x_1,x_2\le 1$ en el rango considerado.
7. El jacobiano (A13) linealiza la dinamica alrededor de cada punto fijo; sus autovalores, traza y determinante determinan la estabilidad local.
8. Cuando las nulclinas interiores coinciden, la interseccion aislada $E$ se convierte en un segmento de equilibrios marginalmente estables.
