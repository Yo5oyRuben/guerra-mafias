# Apéndice: análisis del retrato de fases de la ecuación replicadora de dos variables

En esta sección se reproduce y se explica el comienzo del análisis del retrato de fases de la ecuación replicadora de dos variables.

Las variables dinámicas son

$$
x_1(t), \qquad x_2(t),
$$

donde $x_1$ representa la fracción de cooperadores en la población 1 y $x_2$ representa la fracción de cooperadores en la población 2.

Por tanto,

$$
0\leq x_1\leq 1,
\qquad
0\leq x_2\leq 1.
$$

El objetivo del análisis es estudiar el sistema dinámico definido por

$$
\dot{x}_1=F_1(x_1,x_2),
\qquad
\dot{x}_2=F_2(x_1,x_2).
\qquad \text{(A1)}
$$

A partir de este sistema se determinan sus nulclinas, sus puntos fijos y, posteriormente, la estabilidad de dichos puntos mediante la matriz jacobiana.

## 1. Ecuación replicadora reescalada

Tras el reescalado temporal indicado en el apéndice, las velocidades del sistema son

$$
F_1(x_1,x_2)
=
x_1(1-x_1)
\left\{
\beta[x_1(1-b+r)-r]
+
p[x_2(1-b+\varepsilon)-\varepsilon]
\right\},
$$

$$
F_2(x_1,x_2)
=
x_2(1-x_2)
\left\{
[x_2(1-b+r)-r]
+
\beta p[x_1(1-b+\varepsilon)-\varepsilon]
\right\}.
\qquad \text{(A2)}
$$

Aquí $b$, $r$, $\varepsilon$, $\beta$ y $p$ son parámetros del modelo.

Conviene distinguir entre las velocidades completas $F_1,F_2$ y los factores internos que aparecen entre llaves. Definimos

$$
g_1(x_1,x_2)
=
\beta[x_1(1-b+r)-r]
+
p[x_2(1-b+\varepsilon)-\varepsilon],
$$

$$
g_2(x_1,x_2)
=
[x_2(1-b+r)-r]
+
\beta p[x_1(1-b+\varepsilon)-\varepsilon].
$$

Entonces

$$
F_1=x_1(1-x_1)g_1,
\qquad
F_2=x_2(1-x_2)g_2.
$$

Esta forma es importante porque los factores $x_i(1-x_i)$ hacen que los bordes del cuadrado unidad sean invariantes.

Por ejemplo, si $x_1=0$ o $x_1=1$, entonces

$$
F_1=0
$$

independientemente del valor de $g_1$. De forma análoga, si $x_2=0$ o $x_2=1$, entonces

$$
F_2=0.
$$

```python
import sympy as sp
from IPython.display import display

sp.init_printing()

# Variables dinámicas
x1, x2 = sp.symbols("x_1 x_2", real=True)

# Parámetros del modelo
b, r, eps, beta, p = sp.symbols("b r epsilon beta p", real=True)

# Abreviaturas útiles para seguir la notación del apéndice
M = b - 1 - r          # M = b - 1 - r
N = b - 1 - eps        # N = b - 1 - epsilon

S = beta*r + p*eps     # S = beta*r + p*epsilon
T = r + beta*p*eps     # T = r + beta*p*epsilon

# Factores internos
g1 = beta*(x1*(1 - b + r) - r) + p*(x2*(1 - b + eps) - eps)
g2 = (x2*(1 - b + r) - r) + beta*p*(x1*(1 - b + eps) - eps)

# Velocidades completas
F1 = x1*(1 - x1)*g1
F2 = x2*(1 - x2)*g2

def simp(expr):
    """Simplifica y factoriza una expresión simbólica."""
    return sp.factor(sp.simplify(expr))

def check_zero(label, expr):
    """
    Comprueba si una diferencia simbólica es cero.
    Si la igualdad es correcta, debería mostrarse 0.
    """
    print(label)
    display(simp(expr))

display(F1)
display(F2)
```

## 2. Nulclinas de $F_1=0$

Una nulclina es el conjunto de puntos del plano de fases donde una de las velocidades se anula.

Para $F_1$, imponemos

$$
F_1(x_1,x_2)=0.
$$

Como

$$
F_1=x_1(1-x_1)g_1,
$$

entonces $F_1=0$ si se cumple alguna de las siguientes condiciones:

$$
x_1=0,
\qquad \text{(A3)}
$$

$$
x_1=1,
\qquad \text{(A4)}
$$

o bien

$$
g_1(x_1,x_2)=0.
$$

Ahora derivamos la tercera nulclina. Partimos de

$$
g_1=
\beta[x_1(1-b+r)-r]
+
p[x_2(1-b+\varepsilon)-\varepsilon].
$$

Para simplificar la escritura, definimos

$$
M=b-1-r,
\qquad
N=b-1-\varepsilon.
$$

Entonces

$$
1-b+r=-M,
\qquad
1-b+\varepsilon=-N.
$$

Por tanto,

$$
g_1=
\beta[-Mx_1-r]+p[-Nx_2-\varepsilon].
$$

Es decir,

$$
g_1=-\beta Mx_1-\beta r-pNx_2-p\varepsilon.
$$

Imponiendo $g_1=0$,

$$
-\beta Mx_1-\beta r-pNx_2-p\varepsilon=0.
$$

Agrupando los términos que no contienen $x_2$,

$$
pNx_2=-\beta Mx_1-(\beta r+p\varepsilon).
$$

Por tanto,

$$
x_2=
\frac{-\beta Mx_1-(\beta r+p\varepsilon)}{pN}.
$$

Volviendo a escribir $M=b-1-r$ y $N=b-1-\varepsilon$, obtenemos

$$
x_2=
\frac{
-x_1\beta(b-1-r)-(\beta r+p\varepsilon)
}
{
p(b-1-\varepsilon)
}.
\qquad \text{(A5)}
$$

Así, las nulclinas de $F_1=0$ son las rectas (A3), (A4) y (A5).

## 3. Nulclinas de $F_2=0$

De forma análoga, las nulclinas de $F_2$ se obtienen imponiendo

$$
F_2(x_1,x_2)=0.
$$

Como

$$
F_2=x_2(1-x_2)g_2,
$$

entonces $F_2=0$ si

$$
x_2=0,
\qquad \text{(A6)}
$$

$$
x_2=1,
\qquad \text{(A7)}
$$

o bien

$$
g_2(x_1,x_2)=0.
$$

Derivamos ahora la tercera nulclina. Partimos de

$$
g_2=
[x_2(1-b+r)-r]
+
\beta p[x_1(1-b+\varepsilon)-\varepsilon].
$$

Usando otra vez

$$
1-b+r=-M,
\qquad
1-b+\varepsilon=-N,
$$

tenemos

$$
g_2=(-Mx_2-r)+\beta p(-Nx_1-\varepsilon).
$$

Por tanto,

$$
g_2=-Mx_2-r-\beta pNx_1-\beta p\varepsilon.
$$

Imponiendo $g_2=0$,

$$
-Mx_2-r-\beta pNx_1-\beta p\varepsilon=0.
$$

Despejando $x_2$,

$$
Mx_2=-\beta pNx_1-(r+\beta p\varepsilon),
$$

y por tanto

$$
x_2=
\frac{-\beta pNx_1-(r+\beta p\varepsilon)}{M}.
$$

Volviendo a la notación original,

$$
x_2=
\frac{
-x_1\beta p(b-1-\varepsilon)-(r+\beta p\varepsilon)
}
{
b-1-r
}.
\qquad \text{(A8)}
$$

Así, las nulclinas de $F_2=0$ son las rectas (A6), (A7) y (A8).

```python
# Nulclinas interiores obtenidas resolviendo g1 = 0 y g2 = 0 para x2
A5_obtenida = sp.solve(sp.Eq(g1, 0), x2)[0]
A8_obtenida = sp.solve(sp.Eq(g2, 0), x2)[0]

# Expresiones esperadas según el apéndice
A5_apendice = (-x1*beta*M - S)/(p*N)
A8_apendice = (-x1*beta*p*N - T)/M

print("Nulclina interior de F1=0, ecuación (A5):")
display(sp.Eq(x2, A5_obtenida))

print("Nulclina interior de F2=0, ecuación (A8):")
display(sp.Eq(x2, A8_obtenida))

check_zero("Comprobación de (A5): expresión obtenida - expresión del apéndice =", A5_obtenida - A5_apendice)
check_zero("Comprobación de (A8): expresión obtenida - expresión del apéndice =", A8_obtenida - A8_apendice)
```

## 4. Puntos fijos como intersecciones de nulclinas

Un punto fijo del sistema es un punto $(x_1^\ast,x_2^\ast)$ tal que

$$
F_1(x_1^\ast,x_2^\ast)=0,
\qquad
F_2(x_1^\ast,x_2^\ast)=0.
$$

Por tanto, los puntos fijos se obtienen como cruces entre una nulclina de $F_1=0$ y una nulclina de $F_2=0$.

Las cuatro esquinas del cuadrado unidad son siempre puntos fijos, porque en ellas se anulan simultáneamente los factores $x_i(1-x_i)$.

Así se obtienen los cuatro puntos:

$$
A=(0,1),
$$

$$
B=(1,0),
$$

$$
C=(1,1),
$$

$$
D=(0,0).
$$

El punto $C=(1,1)$ representa cooperación completa en ambas poblaciones, mientras que $D=(0,0)$ representa defection completa en ambas.

Los puntos $A=(0,1)$ y $B=(1,0)$ son estados polarizados: una población está formada solo por defectores y la otra solo por cooperadores.

## 5. Punto $A'$

El punto $A'$ se define como el cruce entre la nulclina

$$
x_1=0
\qquad \text{(A3)}
$$

y la nulclina interior de $F_2=0$,

$$
x_2=
\frac{
-\beta pNx_1-(r+\beta p\varepsilon)
}
{M}.
\qquad \text{(A8)}
$$

Sustituimos $x_1=0$ en (A8):

$$
x_2(A')
=
\frac{
-\beta pN\cdot 0-(r+\beta p\varepsilon)
}
{M}.
$$

Por tanto,

$$
x_2(A')
=
-\frac{r+\beta p\varepsilon}{M}.
$$

Volviendo a escribir $M=b-1-r$, queda

$$
x_1(A')=0,
\qquad
x_2(A')=
-\frac{r+\beta p\varepsilon}{b-1-r}.
\qquad \text{(A9)}
$$

Este punto solo pertenece al cuadrado unidad si

$$
0\leq x_2(A')\leq 1.
$$

## 6. Punto $B'$

El punto $B'$ se define como el cruce entre la nulclina interior de $F_1=0$,

$$
x_2=
\frac{-\beta Mx_1-(\beta r+p\varepsilon)}{pN},
\qquad \text{(A5)}
$$

y la nulclina

$$
x_2=0.
\qquad \text{(A6)}
$$

Sustituimos $x_2=0$ en (A5):

$$
0=
\frac{-\beta Mx_1-(\beta r+p\varepsilon)}{pN}.
$$

Como el denominador no se anula en el caso genérico, debe anularse el numerador:

$$
-\beta Mx_1-(\beta r+p\varepsilon)=0.
$$

Por tanto,

$$
\beta Mx_1=-(\beta r+p\varepsilon),
$$

y

$$
x_1(B')=
-\frac{\beta r+p\varepsilon}{\beta M}.
$$

Volviendo a escribir $M=b-1-r$,

$$
x_2(B')=0,
\qquad
x_1(B')=
-\frac{\beta r+p\varepsilon}{\beta(b-1-r)}.
\qquad \text{(A10)}
$$

Este punto solo pertenece al cuadrado unidad si

$$
0\leq x_1(B')\leq 1.
$$

```python
# Punto A': intersección de x1 = 0 con la nulclina interior de F2
A_prime_obtenido = (sp.Integer(0), simp(A8_obtenida.subs(x1, 0)))
A_prime_apendice = (sp.Integer(0), -T/M)

# Punto B': intersección de la nulclina interior de F1 con x2 = 0
B_prime_x1_obtenido = sp.solve(sp.Eq(A5_obtenida, 0), x1)[0]
B_prime_obtenido = (simp(B_prime_x1_obtenido), sp.Integer(0))
B_prime_apendice = (-S/(beta*M), sp.Integer(0))

print("Punto A' obtenido:")
display(A_prime_obtenido)

print("Punto B' obtenido:")
display(B_prime_obtenido)

check_zero("Comprobación de x2(A'):", A_prime_obtenido[1] - A_prime_apendice[1])
check_zero("Comprobación de x1(B'):", B_prime_obtenido[0] - B_prime_apendice[0])
```

## 7. Punto interior $E$

El punto $E$ se obtiene como el cruce entre las dos nulclinas interiores (A5) y (A8). Equivalentemente, se obtiene resolviendo

$$
g_1=0,
\qquad
g_2=0.
$$

Usamos las abreviaturas

$$
M=b-1-r,
\qquad
N=b-1-\varepsilon,
$$

$$
S=\beta r+p\varepsilon,
\qquad
T=r+\beta p\varepsilon.
$$

A partir de $g_1=0$, habíamos obtenido

$$
-\beta Mx_1-pNx_2-S=0.
$$

Multiplicando por $-1$,

$$
\beta Mx_1+pNx_2=-S.
$$

A partir de $g_2=0$, se obtiene

$$
-\beta pNx_1-Mx_2-T=0.
$$

Multiplicando por $-1$,

$$
\beta pNx_1+Mx_2=-T.
$$

Por tanto, el punto interior $E$ se obtiene resolviendo el sistema lineal

$$
\begin{pmatrix}
\beta M & pN \\
\beta pN & M
\end{pmatrix}
\begin{pmatrix}
x_1\\
x_2
\end{pmatrix}
=
\begin{pmatrix}
-S\\
-T
\end{pmatrix}.
$$

El determinante de la matriz de coeficientes es

$$
\Delta
=
\beta M^2-\beta p^2N^2
=
\beta(M^2-p^2N^2).
$$

Aplicamos ahora la regla de Cramer.

### Coordenada $x_1(E)$

Para calcular $x_1(E)$, sustituimos la primera columna de la matriz de coeficientes por el término independiente:

$$
\Delta_1=
\begin{vmatrix}
-S & pN\\
-T & M
\end{vmatrix}.
$$

Entonces

$$
\Delta_1=(-S)M-pN(-T).
$$

Por tanto,

$$
\Delta_1=-MS+pNT.
$$

Así,

$$
x_1(E)=\frac{\Delta_1}{\Delta}
=
\frac{-MS+pNT}{\beta(M^2-p^2N^2)}.
$$

Multiplicando numerador y denominador por $-1$, queda

$$
x_1(E)=
\frac{MS-pNT}{\beta(p^2N^2-M^2)}.
$$

Sustituyendo

$$
M=b-1-r,
\qquad
N=b-1-\varepsilon,
$$

$$
S=\beta r+p\varepsilon,
\qquad
T=r+\beta p\varepsilon,
$$

obtenemos

$$
x_1(E)=
\frac{
(b-1-r)(\beta r+p\varepsilon)
-
p(b-1-\varepsilon)(r+\beta p\varepsilon)
}
{
\beta\left[p^2(b-1-\varepsilon)^2-(b-1-r)^2\right]
}.
\qquad \text{(A11)}
$$

### Coordenada $x_2(E)$

Para calcular $x_2(E)$, sustituimos la segunda columna de la matriz de coeficientes por el término independiente:

$$
\Delta_2=
\begin{vmatrix}
\beta M & -S\\
\beta pN & -T
\end{vmatrix}.
$$

Entonces

$$
\Delta_2=\beta M(-T)-(-S)\beta pN.
$$

Por tanto,

$$
\Delta_2=-\beta MT+\beta pNS.
$$

Dividiendo por el determinante total,

$$
x_2(E)=
\frac{\Delta_2}{\Delta}
=
\frac{-\beta MT+\beta pNS}{\beta(M^2-p^2N^2)}.
$$

Cancelando el factor común $\beta$,

$$
x_2(E)=
\frac{-MT+pNS}{M^2-p^2N^2}.
$$

Multiplicando numerador y denominador por $-1$,

$$
x_2(E)=
\frac{MT-pNS}{p^2N^2-M^2}.
$$

Sustituyendo las definiciones de $M,N,S,T$, queda

$$
x_2(E)=
\frac{
(b-1-r)(r+\beta p\varepsilon)
-
p(b-1-\varepsilon)(\beta r+p\varepsilon)
}
{
p^2(b-1-\varepsilon)^2-(b-1-r)^2
}.
\qquad \text{(A12)}
$$

El punto $E$ solo es un punto fijo interior si

$$
0<x_1(E)<1,
\qquad
0<x_2(E)<1.
$$

Además, si

$$
p^2(b-1-\varepsilon)^2-(b-1-r)^2=0,
$$

la expresión anterior no es válida como solución aislada. Ese caso corresponde a una situación degenerada de las nulclinas interiores.

```python
# Resolución simbólica directa de g1 = 0, g2 = 0
sol_E = sp.solve([sp.Eq(g1, 0), sp.Eq(g2, 0)], [x1, x2], dict=True)[0]

x1_E_obtenido = sol_E[x1]
x2_E_obtenido = sol_E[x2]

# Expresiones esperadas según (A11)-(A12)
x1_E_apendice = (M*S - p*N*T)/(beta*(p**2*N**2 - M**2))
x2_E_apendice = (M*T - p*N*S)/(p**2*N**2 - M**2)

print("x1(E) obtenido:")
display(sp.factor(x1_E_obtenido))

print("x2(E) obtenido:")
display(sp.factor(x2_E_obtenido))

check_zero("Comprobación de x1(E), ecuación (A11):", x1_E_obtenido - x1_E_apendice)
check_zero("Comprobación de x2(E), ecuación (A12):", x2_E_obtenido - x2_E_apendice)
```

## 8. Cruces omitidos

En principio, cada nulclina de $F_1=0$ puede cruzarse con cada nulclina de $F_2=0$.

Como hay tres nulclinas para $F_1$ y tres para $F_2$, hay nueve cruces candidatos.

Sin embargo, algunos de esos cruces quedan fuera del cuadrado unidad en el rango de parámetros considerado.

Los cruces omitidos en el apéndice son:

$$
(A4)-(A8),
$$

es decir, el cruce de

$$
x_1=1
$$

con la nulclina interior de $F_2=0$, y

$$
(A5)-(A7),
$$

es decir, el cruce de la nulclina interior de $F_1=0$ con

$$
x_2=1.
$$

Veamos las expresiones que se obtienen para esos cruces.

### Cruce $(A4)-(A8)$

El cruce $(A4)-(A8)$ se obtiene tomando la nulclina

$$
x_1=1
$$

y sustituyéndola en la nulclina interior de $F_2=0$,

$$
x_2=
\frac{-\beta pN x_1-T}{M},
$$

donde hemos definido

$$
M=b-1-r,
\qquad
N=b-1-\varepsilon,
\qquad
T=r+\beta p\varepsilon.
$$

Sustituyendo $x_1=1$,

$$
x_2=
\frac{-\beta pN-T}{M}.
$$

Usando las definiciones de $N$ y $T$,

$$
x_2=
\frac{-\beta p(b-1-\varepsilon)-(r+\beta p\varepsilon)}
{b-1-r}.
$$

Desarrollamos el numerador:

$$
-\beta p(b-1-\varepsilon)-(r+\beta p\varepsilon)
=
-\beta p(b-1)+\beta p\varepsilon-r-\beta p\varepsilon.
$$

Los términos $\beta p\varepsilon$ y $-\beta p\varepsilon$ se cancelan, de modo que

$$
-\beta p(b-1-\varepsilon)-(r+\beta p\varepsilon)
=
-\beta p(b-1)-r.
$$

Por tanto, el cruce tiene coordenadas

$$
\left(
1,
-\frac{\beta p(b-1)+r}{b-1-r}
\right).
$$

Ahora comprobamos si la segunda coordenada puede pertenecer al intervalo físico

$$
0\leq x_2\leq 1.
$$

En el rango de parámetros del apéndice se considera

$$
b>1,
\qquad
0\leq r\leq 1,
\qquad
\beta\geq 1,
\qquad
0\leq p\leq 1.
$$

Por tanto,

$$
\beta p(b-1)+r\geq 0.
$$

Distinguimos dos casos.

#### Caso 1: $b-1-r>0$

Si

$$
b-1-r>0,
$$

entonces el denominador es positivo. Como el numerador $\beta p(b-1)+r$ es no negativo y aparece con un signo menos,

$$
x_2=
-\frac{\beta p(b-1)+r}{b-1-r}
\leq 0.
$$

En el caso genérico, será estrictamente negativo. Por tanto, el punto queda por debajo del cuadrado unidad y no es un punto físico.

#### Caso 2: $b-1-r<0$

Si

$$
b-1-r<0,
$$

entonces

$$
r-(b-1)>0.
$$

Podemos reescribir

$$
x_2=
-\frac{\beta p(b-1)+r}{b-1-r}
=
\frac{\beta p(b-1)+r}{r-(b-1)}.
$$

Ahora comparamos numerador y denominador. Como $b-1>0$, se cumple

$$
r>r-(b-1).
$$

Además,

$$
\beta p(b-1)\geq 0.
$$

Por tanto,

$$
\beta p(b-1)+r>r-(b-1).
$$

Como el denominador $r-(b-1)$ es positivo, se obtiene

$$
x_2>1.
$$

Así, en este caso el punto queda por encima del cuadrado unidad.

En conclusión, el cruce $(A4)-(A8)$ queda fuera del cuadrado unidad: si $b-1-r>0$, entonces $x_2<0$; si $b-1-r<0$, entonces $x_2>1$. El caso $b-1-r=0$ es degenerado porque el denominador se anula y debe tratarse por separado.

### Cruce $(A5)-(A7)$

El cruce $(A5)-(A7)$ se obtiene tomando la nulclina

$$
x_2=1
$$

y sustituyéndola en la nulclina interior de $F_1=0$,

$$
x_2=
\frac{-\beta Mx_1-S}{pN},
$$

donde

$$
M=b-1-r,
\qquad
N=b-1-\varepsilon,
\qquad
S=\beta r+p\varepsilon.
$$

Sustituyendo $x_2=1$,

$$
1=
\frac{-\beta Mx_1-S}{pN}.
$$

Multiplicando por $pN$,

$$
pN=-\beta Mx_1-S.
$$

Despejamos $x_1$:

$$
\beta Mx_1=-(S+pN),
$$

y por tanto

$$
x_1=
-\frac{S+pN}{\beta M}.
$$

Ahora usamos las definiciones de $S$, $N$ y $M$:

$$
S=\beta r+p\varepsilon,
\qquad
N=b-1-\varepsilon,
\qquad
M=b-1-r.
$$

Entonces

$$
S+pN
=
\beta r+p\varepsilon+p(b-1-\varepsilon).
$$

Desarrollando,

$$
S+pN
=
\beta r+p\varepsilon+p(b-1)-p\varepsilon.
$$

Los términos $p\varepsilon$ y $-p\varepsilon$ se cancelan, de modo que

$$
S+pN=\beta r+p(b-1).
$$

Por tanto, el cruce tiene coordenadas

$$
\left(
-\frac{\beta r+p(b-1)}{\beta(b-1-r)},
1
\right).
$$

Ahora comprobamos si la primera coordenada puede pertenecer al intervalo físico

$$
0\leq x_1\leq 1.
$$

En el rango de parámetros considerado en el apéndice,

$$
b>1,
\qquad
0\leq r\leq 1,
\qquad
\beta\geq 1,
\qquad
0\leq p\leq 1.
$$

Por tanto,

$$
\beta r+p(b-1)\geq 0.
$$

Distinguimos dos casos.

#### Caso 1: $b-1-r>0$

Si

$$
b-1-r>0,
$$

entonces el denominador $\beta(b-1-r)$ es positivo. Como el numerador $\beta r+p(b-1)$ es no negativo y aparece con un signo menos,

$$
x_1=
-\frac{\beta r+p(b-1)}{\beta(b-1-r)}
\leq 0.
$$

En el caso genérico, será estrictamente negativo. Por tanto, el punto queda a la izquierda del cuadrado unidad y no es un punto físico.

#### Caso 2: $b-1-r<0$

Si

$$
b-1-r<0,
$$

entonces

$$
r-(b-1)>0.
$$

Podemos reescribir

$$
x_1=
-\frac{\beta r+p(b-1)}{\beta(b-1-r)}
=
\frac{\beta r+p(b-1)}{\beta[r-(b-1)]}.
$$

Ahora comparamos el numerador y el denominador. El numerador es

$$
\beta r+p(b-1),
$$

mientras que el denominador es

$$
\beta[r-(b-1)]=\beta r-\beta(b-1).
$$

La diferencia entre ambos es

$$
\bigl[\beta r+p(b-1)\bigr]
-
\bigl[\beta r-\beta(b-1)\bigr]
=
(p+\beta)(b-1).
$$

Como

$$
p+\beta>0
$$

y

$$
b-1>0,
$$

se cumple

$$
(p+\beta)(b-1)>0.
$$

Por tanto,

$$
\beta r+p(b-1)
>
\beta[r-(b-1)].
$$

Como el denominador $\beta[r-(b-1)]$ es positivo en este caso, se obtiene

$$
x_1>1.
$$

Así, el punto queda a la derecha del cuadrado unidad.

En conclusión, el cruce $(A5)-(A7)$ queda fuera del cuadrado unidad: si $b-1-r>0$, entonces $x_1<0$; si $b-1-r<0$, entonces $x_1>1$. El caso $b-1-r=0$ es degenerado porque el denominador se anula y debe tratarse por separado.

## 9. Coincidencia degenerada de nulclinas interiores

El apéndice señala que puede darse un caso no genérico en el que las dos nulclinas interiores coinciden. En ese caso, en lugar de cortarse en un único punto interior, las dos rectas son la misma recta. Si una parte de esa recta queda dentro del cuadrado unidad, aparece un segmento de equilibrios marginalmente estables.

Las nulclinas interiores son

$$
x_2=
\frac{-\beta Mx_1-S}{pN},
$$

$$
x_2=
\frac{-\beta pNx_1-T}{M},
$$

donde hemos definido

$$
M=b-1-r,
\qquad
N=b-1-\varepsilon,
$$

$$
S=\beta r+p\varepsilon,
\qquad
T=r+\beta p\varepsilon.
$$

Para que dos rectas coincidan, no basta con que se corten en un punto. Deben tener la misma pendiente y la misma ordenada en el origen.

Escribimos ambas nulclinas en forma explícita:

$$
x_2
=
-\frac{\beta M}{pN}x_1-\frac{S}{pN},
$$

$$
x_2
=
-\frac{\beta pN}{M}x_1-\frac{T}{M}.
$$

Por tanto, la primera condición es que las pendientes sean iguales:

$$
-\frac{\beta M}{pN}
=
-\frac{\beta pN}{M}.
$$

Cancelando el signo menos y el factor común $\beta$,

$$
\frac{M}{pN}
=
\frac{pN}{M}.
$$

Multiplicando en cruz,

$$
M^2=p^2N^2.
$$

Por tanto,

$$
M=\pm pN.
$$

Hay dos posibilidades algebraicas. Las analizamos por separado.

### Rama 1: $M=pN$

Tomamos primero

$$
M=pN.
$$

La segunda condición para que las rectas coincidan es que sus ordenadas en el origen sean iguales:

$$
-\frac{S}{pN}
=
-\frac{T}{M}.
$$

Quitando el signo menos,

$$
\frac{S}{pN}
=
\frac{T}{M}.
$$

Multiplicando en cruz,

$$
MS=pNT.
$$

Si ahora usamos $M=pN$, resulta

$$
pNS=pNT.
$$

En el caso genérico $pN\neq 0$, podemos cancelar $pN$:

$$
S=T.
$$

Recordando que

$$
S=\beta r+p\varepsilon,
\qquad
T=r+\beta p\varepsilon,
$$

la condición $S=T$ queda

$$
\beta r+p\varepsilon
=
r+\beta p\varepsilon.
$$

Pasamos todo al mismo lado:

$$
\beta r-r+p\varepsilon-\beta p\varepsilon=0.
$$

Factorizando,

$$
(\beta-1)r-p\varepsilon(\beta-1)=0.
$$

Por tanto,

$$
(\beta-1)(r-p\varepsilon)=0.
$$

En el rango de parámetros del modelo, $\varepsilon<0$, $p\geq 0$ y $r\geq 0$. Por ello, salvo casos degenerados muy particulares,

$$
r-p\varepsilon \neq 0.
$$

Así, la condición relevante es

$$
\beta=1.
$$

Por tanto, para que las nulclinas interiores coincidan en esta rama necesitamos

$$
\beta=1,
\qquad
M=pN.
$$

### Obtención de $b^c$

Usamos ahora la condición

$$
M=pN.
$$

Como

$$
M=b-1-r,
\qquad
N=b-1-\varepsilon,
$$

tenemos

$$
b-1-r=p(b-1-\varepsilon).
$$

Desarrollamos el miembro derecho:

$$
b-1-r=p(b-1)-p\varepsilon.
$$

Agrupamos los términos que contienen $b-1$ en un lado:

$$
(b-1)-p(b-1)=r-p\varepsilon.
$$

Factorizando,

$$
(1-p)(b-1)=r-p\varepsilon.
$$

Si $p\neq 1$, despejamos

$$
b-1=
\frac{r-p\varepsilon}{1-p}.
$$

Por tanto,

$$
b^c=
1+\frac{r-p\varepsilon}{1-p}.
$$

Esta es la expresión indicada en el apéndice para el valor crítico en el que las nulclinas interiores coinciden.

### Rama 2: $M=-pN$

La otra posibilidad algebraica procedente de

$$
M^2=p^2N^2
$$

es

$$
M=-pN.
$$

Veamos qué implicaría.

La igualdad de ordenadas en el origen era

$$
MS=pNT.
$$

Si $M=-pN$, entonces

$$
(-pN)S=pNT.
$$

En el caso genérico $pN\neq 0$, podemos cancelar $pN$:

$$
-S=T.
$$

Es decir,

$$
S+T=0.
$$

Sustituyendo

$$
S=\beta r+p\varepsilon,
\qquad
T=r+\beta p\varepsilon,
$$

obtenemos

$$
\beta r+p\varepsilon+r+\beta p\varepsilon=0.
$$

Agrupando,

$$
(\beta+1)r+p\varepsilon(\beta+1)=0.
$$

Por tanto,

$$
(\beta+1)(r+p\varepsilon)=0.
$$

Como $\beta+1>0$, queda

$$
r=-p\varepsilon.
$$

Por otro lado, la condición $M=-pN$ implica

$$
b-1-r=-p(b-1-\varepsilon).
$$

Desarrollando,

$$
b-1-r=-p(b-1)+p\varepsilon.
$$

Agrupando los términos con $b-1$,

$$
(1+p)(b-1)=r+p\varepsilon.
$$

Pero en esta rama acabamos de obtener

$$
r=-p\varepsilon.
$$

Por tanto,

$$
r+p\varepsilon=0.
$$

Luego

$$
(1+p)(b-1)=0.
$$

Como $1+p>0$, se obtiene

$$
b=1.
$$

Este valor no corresponde al régimen de tentación considerado en el apéndice, donde se estudia la variación para $b>1^+$. Por tanto, esta rama no da la coincidencia relevante para el análisis del retrato de fases.

### Conclusión

La coincidencia no genérica de las dos nulclinas interiores relevante para el apéndice se produce cuando

$$
\beta=1
$$

y

$$
b=b^c=
1+\frac{r-p\varepsilon}{1-p}.
$$

En ese caso, las nulclinas interiores no se cortan en un único punto $E$, sino que coinciden como rectas. Si el tramo correspondiente cae dentro del cuadrado unidad, aparece un segmento de puntos fijos.

Ese segmento se denomina marginalmente estable porque la linealización tiene una dirección en la que no hay retorno ni alejamiento exponencial: el campo de velocidades se anula a lo largo de toda una línea de equilibrios, no solo en un punto aislado.

```python
# Comprobación simbólica de la condición de coincidencia de nulclinas

bc = 1 + (r - p*eps)/(1 - p)

# Primera condición: misma pendiente.
# En la rama relevante, las pendientes coinciden si M = p*N.
condicion_pendiente = (M - p*N).subs(b, bc)

check_zero("Comprobación de M - p*N para b = b^c:", condicion_pendiente)

# Segunda condición: misma ordenada en el origen.
# En la rama M = p*N, esto exige S = T.
condicion_ordenada = (S - T).subs(beta, 1)

check_zero("Comprobación de S - T para beta = 1:", condicion_ordenada)

# Comprobación directa: las dos nulclinas interiores coinciden
# al imponer simultáneamente b = b^c y beta = 1.

nulclina_F1 = (-beta*M*x1 - S)/(p*N)
nulclina_F2 = (-beta*p*N*x1 - T)/M

diferencia_nulclinas = (nulclina_F1 - nulclina_F2).subs({
    b: bc,
    beta: 1
})

check_zero(
    "Comprobación directa: nulclina F1 - nulclina F2 para b=b^c y beta=1:",
    diferencia_nulclinas
)
```

## 10. Matriz jacobiana y linealización

Para estudiar la estabilidad de un punto fijo $x^\ast=(x_1^\ast,x_2^\ast)$, se linealiza el sistema alrededor de dicho punto.

El sistema es

$$
\dot{x}_1=F_1(x_1,x_2),
\qquad
\dot{x}_2=F_2(x_1,x_2).
$$

Escribimos una perturbación pequeña alrededor del punto fijo:

$$
x(t)=x^\ast+\delta x(t),
$$

donde

$$
\delta x(t)=
\begin{pmatrix}
\delta x_1(t)\\
\delta x_2(t)
\end{pmatrix}.
$$

Al expandir el campo de velocidades a primer orden alrededor de $x^\ast$, se obtiene

$$
\dot{\delta x}=J(x^\ast)\delta x,
$$

donde $J(x^\ast)$ es la matriz jacobiana evaluada en el punto fijo.

La matriz jacobiana general es

$$
J(x_1,x_2)
=
\begin{pmatrix}
\dfrac{\partial F_1}{\partial x_1}
&
\dfrac{\partial F_1}{\partial x_2}
\\[1em]
\dfrac{\partial F_2}{\partial x_1}
&
\dfrac{\partial F_2}{\partial x_2}
\end{pmatrix}.
$$

Por tanto, en el punto fijo $x^\ast$,

$$
J(x^\ast)=
\left.
\begin{pmatrix}
\dfrac{\partial F_1}{\partial x_1}
&
\dfrac{\partial F_1}{\partial x_2}
\\[1em]
\dfrac{\partial F_2}{\partial x_1}
&
\dfrac{\partial F_2}{\partial x_2}
\end{pmatrix}
\right|_{x=x^\ast}.
\qquad \text{(A13)}
$$

La estabilidad local del punto fijo se determina estudiando los autovalores de esta matriz. En un sistema bidimensional, también es habitual usar la traza y el determinante del jacobiano evaluado en el punto fijo.

## 11. Derivación explícita del jacobiano

Partimos de la forma

$$
F_1=x_1(1-x_1)g_1,
\qquad
F_2=x_2(1-x_2)g_2.
$$

Recordemos que

$$
g_1=
\beta[x_1(1-b+r)-r]
+
p[x_2(1-b+\varepsilon)-\varepsilon],
$$

$$
g_2=
[x_2(1-b+r)-r]
+
\beta p[x_1(1-b+\varepsilon)-\varepsilon].
$$

Queremos calcular

$$
J(x_1,x_2)=
\begin{pmatrix}
\dfrac{\partial F_1}{\partial x_1}
&
\dfrac{\partial F_1}{\partial x_2}
\\[1em]
\dfrac{\partial F_2}{\partial x_1}
&
\dfrac{\partial F_2}{\partial x_2}
\end{pmatrix}.
$$

Primero derivamos los factores logísticos:

$$
\frac{\partial}{\partial x_1}\left[x_1(1-x_1)\right]
=
1-2x_1,
$$

$$
\frac{\partial}{\partial x_2}\left[x_2(1-x_2)\right]
=
1-2x_2.
$$

Además,

$$
\frac{\partial g_1}{\partial x_1}
=
\beta(1-b+r),
\qquad
\frac{\partial g_1}{\partial x_2}
=
p(1-b+\varepsilon),
$$

y

$$
\frac{\partial g_2}{\partial x_1}
=
\beta p(1-b+\varepsilon),
\qquad
\frac{\partial g_2}{\partial x_2}
=
1-b+r.
$$

Usando la regla del producto,

$$
\frac{\partial F_1}{\partial x_1}
=
(1-2x_1)g_1
+
x_1(1-x_1)\frac{\partial g_1}{\partial x_1}.
$$

Por tanto,

$$
\frac{\partial F_1}{\partial x_1}
=
(1-2x_1)g_1
+
x_1(1-x_1)\beta(1-b+r).
$$

Como el factor $x_1(1-x_1)$ no depende de $x_2$,

$$
\frac{\partial F_1}{\partial x_2}
=
x_1(1-x_1)\frac{\partial g_1}{\partial x_2},
$$

luego

$$
\frac{\partial F_1}{\partial x_2}
=
x_1(1-x_1)p(1-b+\varepsilon).
$$

Análogamente, como $x_2(1-x_2)$ no depende de $x_1$,

$$
\frac{\partial F_2}{\partial x_1}
=
x_2(1-x_2)\frac{\partial g_2}{\partial x_1},
$$

por lo que

$$
\frac{\partial F_2}{\partial x_1}
=
x_2(1-x_2)\beta p(1-b+\varepsilon).
$$

Finalmente, aplicando de nuevo la regla del producto,

$$
\frac{\partial F_2}{\partial x_2}
=
(1-2x_2)g_2
+
x_2(1-x_2)\frac{\partial g_2}{\partial x_2}.
$$

Por tanto,

$$
\frac{\partial F_2}{\partial x_2}
=
(1-2x_2)g_2
+
x_2(1-x_2)(1-b+r).
$$

Así, el jacobiano queda

$$
J(x_1,x_2)=
\begin{pmatrix}
(1-2x_1)g_1+x_1(1-x_1)\beta(1-b+r)
&
x_1(1-x_1)p(1-b+\varepsilon)
\\[1em]
x_2(1-x_2)\beta p(1-b+\varepsilon)
&
(1-2x_2)g_2+x_2(1-x_2)(1-b+r)
\end{pmatrix}.
$$

Esta es la forma explícita del jacobiano que se obtiene a partir de la definición (A13).

```python
# Jacobiano calculado directamente con SymPy a partir de F1 y F2

J_sympy = sp.Matrix([
    [sp.diff(F1, x1), sp.diff(F1, x2)],
    [sp.diff(F2, x1), sp.diff(F2, x2)]
])

# Jacobiano derivado a mano usando:
# F1 = x1(1-x1)g1
# F2 = x2(1-x2)g2

J_manual = sp.Matrix([
    [
        (1 - 2*x1)*g1 + x1*(1 - x1)*beta*(1 - b + r),
        x1*(1 - x1)*p*(1 - b + eps)
    ],
    [
        x2*(1 - x2)*beta*p*(1 - b + eps),
        (1 - 2*x2)*g2 + x2*(1 - x2)*(1 - b + r)
    ]
])

print("Jacobiano calculado directamente con SymPy:")
display(J_sympy)

print("Jacobiano obtenido a partir de la derivación manual:")
display(J_manual)

print("Comprobación entrada a entrada de que ambas matrices coinciden:")
for i in range(2):
    for j in range(2):
        check_zero(f"Entrada ({i+1},{j+1})", J_sympy[i, j] - J_manual[i, j])
```

# Análisis de estabilidad y bifurcaciones

En esta parte seguimos el orden del apéndice:

1. Caso simétrico:
   $$
   N_1=N_2(=N),
   \qquad
   \beta=\frac{N_1}{N_2}=1.
   $$

2. Caso general:
   $$
   N_1\neq N_2,
   \qquad
   \beta=\frac{N_1}{N_2}>1.
   $$

En el paper el castigo interpoblacional se escribe como $\rho<0$. En este notebook mantenemos la notación que veníamos usando,

$$
\varepsilon\equiv \rho<0.
$$

Partimos de la ecuación replicadora ya escrita como

$$
\dot x_1=F_1(x_1,x_2),
\qquad
\dot x_2=F_2(x_1,x_2),
$$

y del jacobiano

$$
J(x_1,x_2)=
\begin{pmatrix}
\dfrac{\partial F_1}{\partial x_1} & \dfrac{\partial F_1}{\partial x_2}\\[0.4em]
\dfrac{\partial F_2}{\partial x_1} & \dfrac{\partial F_2}{\partial x_2}
\end{pmatrix}.
$$

La idea es evaluar este jacobiano en los puntos fijos y estudiar el signo de sus autovalores.

# 1. Caso simétrico: $N_1=N_2(=N)$

En primer lugar el paper considera poblaciones de igual tamaño,

$$
N_1=N_2=N.
$$

Entonces

$$
\beta=\frac{N_1}{N_2}=1.
$$

En este caso las dos poblaciones son idénticas, aunque distinguibles. Por tanto, el retrato de fases debe ser invariante bajo el intercambio

$$
x_1\leftrightarrow x_2.
$$

Esto implica que los estados polarizados

$$
A=(0,1),
\qquad
B=(1,0)
$$

tienen el mismo comportamiento de estabilidad. En particular, pierden estabilidad al mismo valor de $b$.

## 1.1. Jacobiano en las esquinas para $\beta=1$

Las cuatro esquinas del cuadrado unidad son

$$
A=(0,1),
\qquad
B=(1,0),
\qquad
C=(1,1),
\qquad
D=(0,0).
$$

En el caso simétrico basta sustituir $\beta=1$ en el jacobiano general y evaluarlo en estas esquinas.

Como estamos en puntos de borde, el jacobiano queda triangular o diagonal. Por tanto, los autovalores se leen directamente de la diagonal.

```python
# Caso simétrico: beta = 1

J_sym = simp(J_base.subs(beta, 1))

puntos_esquina = {
    "A=(0,1)": {x1: 0, x2: 1},
    "B=(1,0)": {x1: 1, x2: 0},
    "C=(1,1)": {x1: 1, x2: 1},
    "D=(0,0)": {x1: 0, x2: 0},
}

for nombre, punto in puntos_esquina.items():
    print("\n" + nombre)
    J_punto = simp(J_sym.subs(punto))
    display(J_punto)
    print("Autovalores:")
    display([simp(ev) for ev in J_punto.eigenvals().keys()])
```

## 1.2. Estabilidad de $D$ y valor crítico $r_c$

En el punto completamente defectivo

$$
D=(0,0),
$$

al tomar $\beta=1$ se obtiene

$$
J(D)=
\begin{pmatrix}
-(r+p\varepsilon) & 0\\
0 & -(r+p\varepsilon)
\end{pmatrix}.
$$

Por tanto,

$$
\lambda_1(D)=\lambda_2(D)=-(r+p\varepsilon).
$$

El cambio de estabilidad ocurre cuando

$$
r+p\varepsilon=0.
$$

Así,

$$
r_c=-p\varepsilon.
$$

Como $\varepsilon<0$, este valor crítico es positivo.

Entonces:

- si $r>r_c=-p\varepsilon$, el punto $D$ es estable;
- si $r<r_c=-p\varepsilon$, el punto $D$ es inestable.

Esta es la separación entre los dos escenarios simétricos del apéndice.

```python
# Valor crítico r_c en el caso simétrico

r_c_sym = -p*eps

lambda_D_sym = -(r + p*eps)

display(sp.Eq(sp.Symbol("lambda_D"), lambda_D_sym))
display(sp.Eq(sp.Symbol("r_c"), r_c_sym))

check_zero(
    "Comprobación: lambda_D se anula en r = r_c",
    lambda_D_sym.subs(r, r_c_sym)
)
```

## 1.3. Pérdida de estabilidad de $A$ y $B$: $b^{\mathrm{up}}$

En el caso simétrico, los puntos

$$
A=(0,1),
\qquad
B=(1,0)
$$

son equivalentes por intercambio de poblaciones.

Al evaluar el jacobiano en $A$, el autovalor que cambia de signo es

$$
\lambda_A=b-1+p\varepsilon.
$$

En $B$ aparece el mismo valor crítico. La bifurcación ocurre cuando

$$
b-1+p\varepsilon=0.
$$

Por tanto,

$$
b^{\mathrm{up}}=1-p\varepsilon.
$$

Este es el valor en el que los estados polarizados pierden estabilidad.

```python
# Valor b^up en el caso simétrico

b_up_sym = 1 - p*eps

lambda_A_sym = b - 1 + p*eps
lambda_B_sym = b - 1 + p*eps

display(sp.Eq(sp.Symbol("b^up"), b_up_sym))

check_zero(
    "Comprobación en A: lambda_A se anula en b = b^up",
    lambda_A_sym.subs(b, b_up_sym)
)

check_zero(
    "Comprobación en B: lambda_B se anula en b = b^up",
    lambda_B_sym.subs(b, b_up_sym)
)
```

## 1.4. Bifurcación de los cuasipolarizados: $b^c$

En el escenario simétrico con

$$
r<r_c=-p\varepsilon,
$$

los puntos polarizados $A$ y $B$ dejan de ser atractores en $b^{\mathrm{up}}$ y aparecen los estados cuasipolarizados

$$
A',
\qquad
B'.
$$

Estos estados se desestabilizan cuando las nulclinas interiores coinciden. En el caso simétrico, el paper obtiene

$$
b^c
=
1+\frac{r-p\varepsilon}{1-p}.
$$

En ese valor aparece un segmento de equilibrios marginalmente estables. Para

$$
b>b^c
$$

el atractor global pasa a ser el punto interior simétrico $E$.

```python
# Valor b^c en el caso simétrico

b_c_sym = 1 + (r - p*eps)/(1 - p)

display(sp.Eq(sp.Symbol("b^c"), b_c_sym))

# Comprobación con las condiciones de bifurcación generales:
# b_A^c y b_B^c deben coincidir al poner beta=1.

b_c_A_general = 1 + beta*(r**2 - (p*eps)**2)/((beta*r + p*eps) - p*(r + beta*p*eps))
b_c_B_general = 1 + (r**2 - (p*eps)**2)/((r + beta*p*eps) - p*(beta*r + p*eps))

check_zero(
    "Comprobación: b_A^c(beta=1) - b^c",
    b_c_A_general.subs(beta, 1) - b_c_sym
)

check_zero(
    "Comprobación: b_B^c(beta=1) - b^c",
    b_c_B_general.subs(beta, 1) - b_c_sym
)
```

## 1.5. Secuencia de bifurcaciones en el caso simétrico

Una vez obtenidos los valores críticos

$$
r_c=-p\varepsilon,
\qquad
b^{\mathrm{up}}=1-p\varepsilon,
\qquad
b^c=1+\frac{r-p\varepsilon}{1-p},
$$

podemos ordenar los cambios cualitativos del retrato de fases cuando aumenta la tentación $b$ desde $1^+$.

El papel de $b^{\mathrm{up}}$ y $b^c$ es distinto:

- $b^{\mathrm{up}}$ marca la pérdida de estabilidad de los estados polarizados $A$ y $B$.
- $b^c$ marca la pérdida de estabilidad de los estados cuasipolarizados $A'$ y $B'$ mediante la coincidencia de las nulclinas interiores.

Por tanto, $b^{\mathrm{up}}$ no se interpreta como coincidencia de nulclinas, sino como el valor en el que $A'$ y $B'$ entran en el cuadrado unidad por los vértices $A$ y $B$.

En efecto, para $\beta=1$,

$$
A'=\left(0,-\frac{r+p\varepsilon}{b-1-r}\right),
\qquad
B'=\left(-\frac{r+p\varepsilon}{b-1-r},0\right).
$$

El punto $A'$ entra por $A=(0,1)$ cuando

$$
x_2(A')=1.
$$

Esto da

$$
-\frac{r+p\varepsilon}{b-1-r}=1,
$$

y por tanto

$$
b=1-p\varepsilon=b^{\mathrm{up}}.
$$

Análogamente, $B'$ entra por $B=(1,0)$ en el mismo valor, debido a la simetría de intercambio entre poblaciones.

El apéndice distingue entonces dos escenarios según la estabilidad de $D$.

---

### Escenario (s1): $r>r_c=-p\varepsilon$

En este caso,

$$
r+p\varepsilon>0,
$$

y los autovalores de $D$ son negativos:

$$
\lambda_1(D)=\lambda_2(D)=-(r+p\varepsilon)<0.
$$

Por tanto, $D$ es estable.

Para

$$
1<b<b^{\mathrm{up}},
$$

coexisten tres atractores:

$$
D,\qquad A,\qquad B.
$$

Al llegar a

$$
b=b^{\mathrm{up}},
$$

los estados polarizados $A$ y $B$ colisionan con los estados cuasipolarizados $A'$ y $B'$. A partir de ese valor, $A$ y $B$ dejan de ser atractores y pasan a ser puntos silla.

Como $D$ ya era estable, para

$$
b>b^{\mathrm{up}}
$$

el único atractor global es $D$.

La secuencia de atractores es

$$
D,A,B
\xrightarrow{\ b^{\mathrm{up}}\ }
D.
\qquad \text{(A14)}
$$

---

### Escenario (s2): $r<r_c=-p\varepsilon$

En este caso,

$$
r+p\varepsilon<0,
$$

y los autovalores de $D$ son positivos:

$$
\lambda_1(D)=\lambda_2(D)=-(r+p\varepsilon)>0.
$$

Por tanto, $D$ es inestable.

Para

$$
1<b<b^{\mathrm{up}},
$$

los atractores son los estados polarizados

$$
A,
\qquad
B.
$$

En

$$
b=b^{\mathrm{up}},
$$

los estados cuasipolarizados $A'$ y $B'$ entran en el cuadrado unidad por los vértices $A$ y $B$. Desde ese momento, los atractores dejan de ser exactamente polarizados y pasan a ser cuasipolarizados.

Así, para

$$
b^{\mathrm{up}}<b<b^c,
$$

los atractores son

$$
A',
\qquad
B'.
$$

Finalmente, en

$$
b=b^c,
$$

las dos nulclinas interiores coinciden. En ese instante aparece un segmento de equilibrios marginalmente estables que une $A'$ y $B'$.

Para

$$
b>b^c,
$$

la dinámica queda atraída por el equilibrio interior simétrico $E$.

La secuencia completa es

$$
A,B
\xrightarrow{\ b^{\mathrm{up}}\ }
A',B'
\xrightarrow{\ b^c\ }
E.
\qquad \text{(A15)}
$$

### Comentario: posición de $A'$, $B'$ y ausencia de $E$ en el escenario (s1)

Conviene aclarar qué son exactamente los puntos $A'$ y $B'$ y por qué no siempre aparecen como atractores físicos.

Los puntos $A'$ y $B'$ son equilibrios de borde. En el caso simétrico,

$$
N_1=N_2,
\qquad
\beta=1,
$$

sus coordenadas son

$$
A'=\left(0,-\frac{r+p\varepsilon}{b-1-r}\right),
$$

$$
B'=\left(-\frac{r+p\varepsilon}{b-1-r},0\right).
$$

Por simetría, ambos dependen de la misma coordenada

$$
y(b)=-\frac{r+p\varepsilon}{b-1-r}.
$$

Así,

$$
A'=(0,y),
\qquad
B'=(y,0).
$$

Para que estos puntos estén dentro del cuadrado unidad debe cumplirse

$$
0\leq y\leq 1.
$$

En el valor

$$
b^{\mathrm{up}}=1-p\varepsilon,
$$

tenemos

$$
y(b^{\mathrm{up}})
=
-\frac{r+p\varepsilon}{b^{\mathrm{up}}-1-r}
=
-\frac{r+p\varepsilon}{-p\varepsilon-r}
=
-\frac{r+p\varepsilon}{-(r+p\varepsilon)}
=
1.
$$

Por tanto, en $b=b^{\mathrm{up}}$ se cumple

$$
A'=A=(0,1),
\qquad
B'=B=(1,0).
$$

Es decir, $A'$ y $B'$ están exactamente sobre el borde del cuadrado, coincidiendo con los estados polarizados. Lo que ocurre al aumentar un poco más $b$ depende del signo de

$$
r+p\varepsilon.
$$

---

#### Caso $r<r_c=-p\varepsilon$

Si

$$
r<r_c=-p\varepsilon,
$$

entonces

$$
r+p\varepsilon<0.
$$

Para $b>b^{\mathrm{up}}$, la coordenada $y(b)$ queda entre 0 y 1:

$$
0<y(b)<1.
$$

Por tanto, $A'$ y $B'$ entran en el cuadrado unidad. En este caso sí aparecen como atractores físicos y sustituyen a los estados polarizados $A$ y $B$.

Por eso en el escenario (s2) aparece la transición

$$
A,B
\xrightarrow{\ b^{\mathrm{up}}\ }
A',B'.
$$

---

#### Caso $r>r_c=-p\varepsilon$

Si

$$
r>r_c=-p\varepsilon,
$$

entonces

$$
r+p\varepsilon>0.
$$

En $b=b^{\mathrm{up}}$, $A'$ y $B'$ coinciden con $A$ y $B$, pero al aumentar $b$ no entran en el interior del cuadrado como atractores. Al contrario, salen de la región física $0\leq x_1,x_2\leq 1$.

Por eso, en el escenario (s1), al llegar a $b^{\mathrm{up}}$ los estados $A$ y $B$ pierden estabilidad, pero no son sustituidos por una fase atractora $A',B'$. Como $D$ ya era estable, el sistema queda finalmente atraído por $D$.

De ahí la secuencia

$$
D,A,B
\xrightarrow{\ b^{\mathrm{up}}\ }
D.
$$

---

#### ¿Por qué no aparece $E$ ni una bifurcación física en $b^c$ en el escenario (s1)?

El punto $E$ es el cruce de las dos nulclinas interiores. En el caso simétrico tiene la forma

$$
E=(x_E,x_E).
$$

A partir de las coordenadas generales de $E$, tomando $\beta=1$, se obtiene

$$
x_E=
\frac{(r+p\varepsilon)\left[(b-1-r)-p(b-1-\varepsilon)\right]}
{p(b-1-\varepsilon)^2-(b-1-r)^2}.
$$

Factorizando el denominador,

$$
p(b-1-\varepsilon)^2-(b-1-r)^2
=
\left[p(b-1-\varepsilon)-(b-1-r)\right]
\left[p(b-1-\varepsilon)+(b-1-r)\right].
$$

Como el numerador contiene el factor opuesto,

$$
(b-1-r)-p(b-1-\varepsilon)
=
-\left[p(b-1-\varepsilon)-(b-1-r)\right],
$$

queda

$$
x_E=
-\frac{r+p\varepsilon}
{(b-1-r)+p(b-1-\varepsilon)}.
$$

Es decir,

$$
x_E=
-\frac{r+p\varepsilon}
{(1+p)(b-1)-(r+p\varepsilon)}.
$$

Ahora consideramos el escenario (s1):

$$
r>r_c=-p\varepsilon.
$$

Entonces

$$
r+p\varepsilon>0.
$$

Llamando

$$
a=r+p\varepsilon>0,
$$

se tiene

$$
x_E=
-\frac{a}{(1+p)(b-1)-a}.
$$

Para que $E$ estuviera dentro del cuadrado unidad necesitaríamos

$$
0\leq x_E\leq 1.
$$

Pero esto no ocurre:

- si el denominador es positivo, entonces $x_E<0$;
- si el denominador es negativo, entonces $x_E>1$.

Por tanto, en el escenario (s1), $E$ está siempre fuera del cuadrado unidad.

Esto explica por qué no aparece la bifurcación en $b^c$ dentro de la secuencia (s1). Aunque algebraicamente se pueda escribir

$$
b^c=
1+\frac{r-p\varepsilon}{1-p},
$$

la coincidencia de nulclinas correspondiente no produce una transición física dentro del cuadrado unidad. Por eso el único cambio relevante en (s1) es la pérdida de estabilidad de $A$ y $B$ en $b^{\mathrm{up}}$.

En cambio, cuando

$$
r<r_c=-p\varepsilon,
$$

el punto $E$ sí pertenece al cuadrado unidad. En ese caso, $E$ existe como equilibrio físico y se convierte en atractor global al pasar $b^c$. Por eso en el escenario (s2) sí aparece la secuencia

$$
A,B
\xrightarrow{\ b^{\mathrm{up}}\ }
A',B'
\xrightarrow{\ b^c\ }
E.
$$

### Interpretación física del valor crítico $r_c$

El valor crítico

$$
r_c=-p\varepsilon
$$

separa los regímenes en los que el estado completamente defectivo

$$
D=(0,0)
$$

es inestable o estable.

Recordemos que $x_1$ y $x_2$ representan las fracciones de cooperadores en las poblaciones 1 y 2. Por tanto,

$$
D=(0,0)
$$

significa

$$
x_1=0,
\qquad
x_2=0,
$$

es decir, ninguna de las dos poblaciones tiene cooperadores. Todos los individuos de ambas poblaciones son defectores.

En el caso simétrico, los autovalores de $D$ son

$$
\lambda_1(D)=\lambda_2(D)=-(r+p\varepsilon).
$$

Por tanto, si

$$
r<-p\varepsilon,
$$

entonces

$$
r+p\varepsilon<0,
$$

y se tiene

$$
\lambda_1(D)=\lambda_2(D)>0.
$$

En consecuencia, $D$ es inestable frente a pequeñas fluctuaciones cooperativas. Es decir, si aparece una pequeña fracción de cooperadores en alguna de las poblaciones, la dinámica se aleja de la deserción completa.

En cambio, si

$$
r>-p\varepsilon,
$$

entonces

$$
r+p\varepsilon>0,
$$

y por tanto

$$
\lambda_1(D)=\lambda_2(D)<0.
$$

En este caso, $D$ es estable: si el sistema está cerca de la deserción completa, vuelve hacia ella.

Esta condición puede interpretarse como un balance entre dos efectos. Por un lado, $r$ mide el excedente asociado a las interacciones defectivas dentro de cada población. Por otro lado, el término $p\varepsilon$, con $\varepsilon<0$, representa el castigo medio asociado a las interacciones defectivas entre poblaciones.

Cuando el castigo interpoblacional domina,

$$
r<-p\varepsilon,
$$

la deserción completa se vuelve inestable. Cuando domina el excedente intrapoblacional,

$$
r>-p\varepsilon,
$$

la deserción completa puede ser estable.

Sin embargo, incluso en este segundo caso, los estados polarizados

$$
A=(0,1),
\qquad
B=(1,0)
$$

pueden seguir siendo atractores para valores bajos de $b$. Esto significa que la estabilidad de $D$ no elimina necesariamente la posibilidad de acabar en estados polarizados. El resultado final depende también de la condición inicial y de la cuenca de atracción correspondiente.

En resumen, $r_c$ no solo marca un cambio algebraico en los autovalores de $D$, sino que separa dos comportamientos físicos distintos: deserción completa inestable frente a fluctuaciones cooperativas, o deserción completa estable.

# 2. Caso general: $N_1\neq N_2$

Después del caso simétrico, el paper estudia el caso general

$$
N_1\neq N_2.
$$

Sin pérdida de generalidad se toma

$$
\beta=\frac{N_1}{N_2}>1.
$$

Esto significa que la población 1 es la población grande:

$$
N_1>N_2.
$$

Recordemos que $x_i$ representa la fracción de cooperadores en la población $i$. Por tanto,

$$
A=(0,1)
$$

significa que

$$
x_1=0,\qquad x_2=1.
$$

Es decir, en $A$ la población 1 es completamente defectiva y la población 2 es completamente cooperativa. Como $N_1>N_2$, en este estado la población defectiva es la de mayor tamaño.

En cambio,

$$
B=(1,0)
$$

significa que

$$
x_1=1,\qquad x_2=0.
$$

Es decir, en $B$ la población 1 es completamente cooperativa y la población 2 es completamente defectiva. Por tanto, en este estado la población defectiva es la pequeña.

La diferencia importante respecto al caso simétrico es que ahora ya no existe simetría de intercambio entre las poblaciones:

$$
x_1\leftrightarrow x_2.
$$

Por eso los estados $A$ y $B$ dejan de ser equivalentes.

Más adelante veremos, mediante el análisis de estabilidad, que esta asimetría hace que el estado $A$ sea más robusto que el estado $B$.

B=(1,0),

C=(1,1),
\qquad

$$

Ahora mantenemos $\beta$ como parámetro libre.

Los autovalores relevantes son:

\lambda_1(D)=-(\beta r+p\varepsilon),

\lambda_2(D)=-(r+\beta p\varepsilon).

### Punto $A$

\lambda_1(A)=-\beta r-p(b-1),

\lambda_1(C)=(\beta+p)(b-1),

Como $b>1$, el punto $C$ es siempre inestable.

## 2.2. Valores críticos de $D$: $r_c^A$ y $r_c^B$

El punto completamente defectivo es

$$
D=(0,0).
$$

Sus autovalores son

$$
\lambda_1(D)=-(\beta r+p\varepsilon),
$$

$$
\lambda_2(D)=-(r+\beta p\varepsilon).
$$

Los cambios de estabilidad ocurren cuando alguno de ellos se anula.

De

$$
r+\beta p\varepsilon=0
$$

obtenemos

$$
r_c^A=-\beta p\varepsilon.
$$

De

$$
\beta r+p\varepsilon=0
$$

obtenemos

$$
r_c^B=-\frac{p\varepsilon}{\beta}.
$$

Como

$$
\varepsilon<0,
\qquad
\beta>1,
$$

se cumple

$$
r_c^B<r_c^A.
$$

Por tanto:

- si $r>r_c^A$, entonces $D$ es estable;
- si $r_c^B<r<r_c^A$, entonces $D$ es una silla;
- si $r<r_c^B$, entonces $D$ es una fuente.

```python
# Valores críticos de r para el estado D

r_c_A = -beta*p*eps
r_c_B = -p*eps/beta

display(sp.Eq(sp.Symbol("r_c^A"), r_c_A))
display(sp.Eq(sp.Symbol("r_c^B"), r_c_B))

lambda_D_1 = -(beta*r + p*eps)
lambda_D_2 = -(r + beta*p*eps)

check_zero(
    "lambda_D_2 se anula en r = r_c^A",
    lambda_D_2.subs(r, r_c_A)
)

check_zero(
    "lambda_D_1 se anula en r = r_c^B",
    lambda_D_1.subs(r, r_c_B)
)
```

## 2.3. Pérdida de estabilidad de $A$ y $B$

En el caso general, $A$ y $B$ ya no pierden estabilidad al mismo valor de $b$, porque al tomar

$$
\beta=\frac{N_1}{N_2}>1
$$

se ha roto la simetría de intercambio entre poblaciones.

Para estudiar cuándo pierden estabilidad, evaluamos el jacobiano en cada punto fijo. Como $A$ y $B$ están en el borde del cuadrado unidad, el jacobiano queda triangular, de modo que los autovalores se leen directamente de la diagonal.

### Estado $A=(0,1)$

En el estado

$$
A=(0,1),
$$

la población 1 es completamente defectiva y la población 2 es completamente cooperativa. Los autovalores del jacobiano en este punto son

$$
\lambda_1(A)=-\beta r-p(b-1),
$$

$$
\lambda_2(A)=b-1+\beta p\varepsilon.
$$

En el intervalo de parámetros considerado,

$$
b>1,
\qquad
r\geq 0,
\qquad
p\geq 0,
\qquad
\beta>1,
$$

el primer autovalor es siempre negativo:

$$
\lambda_1(A)=-\beta r-p(b-1)<0.
$$

Por tanto, este autovalor no puede producir la pérdida de estabilidad de $A$.

El autovalor relevante es entonces

$$
\lambda_2(A)=b-1+\beta p\varepsilon,
$$

porque sí puede cambiar de signo al variar $b$, ya que $\varepsilon<0$.

La bifurcación ocurre cuando

$$
\lambda_2(A)=0.
$$

Es decir,

$$
b-1+\beta p\varepsilon=0.
$$

Despejando,

$$
b_A^{\mathrm{up}}=1-\beta p\varepsilon.
\qquad \text{(A17)}
$$

### Estado $B=(1,0)$

En el estado

$$
B=(1,0),
$$

la población 1 es completamente cooperativa y la población 2 es completamente defectiva. Los autovalores del jacobiano en este punto son

$$
\lambda_1(B)=\beta(b-1)+p\varepsilon,
$$

$$
\lambda_2(B)=-r-\beta p(b-1).
$$

De nuevo, uno de ellos es siempre negativo en el intervalo de parámetros considerado:

$$
\lambda_2(B)=-r-\beta p(b-1)<0.
$$

Por tanto, este autovalor no puede causar la pérdida de estabilidad de $B$.

El autovalor relevante es

$$
\lambda_1(B)=\beta(b-1)+p\varepsilon,
$$

porque es el que puede cambiar de signo cuando aumenta $b$.

La bifurcación ocurre cuando

$$
\lambda_1(B)=0.
$$

Es decir,

$$
\beta(b-1)+p\varepsilon=0.
$$

Despejando,

$$
b_B^{\mathrm{up}}=1-\frac{p\varepsilon}{\beta}.
\qquad \text{(A16)}
$$

Como

$$
\varepsilon<0,
\qquad
\beta>1,
$$

se cumple

$$
b_B^{\mathrm{up}}<b_A^{\mathrm{up}}.
$$

Por tanto, al aumentar $b$ desde $1^+$, el estado $B$ pierde estabilidad antes que el estado $A$.

Esto refleja la asimetría entre las dos poblaciones: en el caso general, $A$ y $B$ ya no son equivalentes, y el estado $A$, donde la población defectiva es la de mayor tamaño, resulta más robusto que $B$.

```python
# Bifurcaciones de los estados polarizados A y B

b_up_A = 1 - beta*p*eps
b_up_B = 1 - p*eps/beta

display(sp.Eq(sp.Symbol("b_A^up"), b_up_A))
display(sp.Eq(sp.Symbol("b_B^up"), b_up_B))

lambda_A_relevante = b - 1 + beta*p*eps
lambda_B_relevante = beta*(b - 1) + p*eps

check_zero(
    "lambda_A se anula en b = b_A^up",
    lambda_A_relevante.subs(b, b_up_A)
)

check_zero(
    "lambda_B se anula en b = b_B^up",
    lambda_B_relevante.subs(b, b_up_B)
)
```

## 2.4. Estados cuasipolarizados $A'$ y $B'$

Además de los estados completamente polarizados

$$
A=(0,1),
\qquad
B=(1,0),
$$

el apéndice introduce dos puntos fijos de borde, llamados estados cuasipolarizados:

$$
A',
\qquad
B'.
$$

Se llaman cuasipolarizados porque una de las dos poblaciones queda completamente defectiva, mientras que la otra no tiene por qué estar exactamente en $0$ o en $1$, sino que puede tener una fracción intermedia de cooperadores.

Por ahora los tratamos como **candidatos** a puntos fijos físicos. Es decir, primero obtenemos sus coordenadas como intersecciones de nulclinas. Después habrá que comprobar en qué condiciones pertenecen al cuadrado unidad y cuándo son estables.

---

### Estado $A'$

El punto $A'$ aparece sobre el borde

$$
x_1=0.
$$

Por tanto, en $A'$ la población 1 es completamente defectiva. La población 2, en cambio, puede tener una fracción intermedia de cooperadores. Escribimos entonces

$$
A'=(0,x_2(A')).
$$

Este punto se obtiene como el cruce entre la nulclina

$$
x_1=0
$$

y la nulclina interior de $F_2=0$.

De la primera parte del apéndice teníamos

$$
x_2=
\frac{-\beta pN x_1-T}{M},
$$

donde

$$
M=b-1-r,
\qquad
N=b-1-\varepsilon,
\qquad
T=r+\beta p\varepsilon.
$$

Sustituyendo $x_1=0$, queda

$$
x_2(A')
=
-\frac{T}{M}.
$$

Por tanto,

$$
A'=
\left(
0,
-\frac{T}{M}
\right).
$$

Es decir,

$$
A'=
\left(
0,
-\frac{r+\beta p\varepsilon}{b-1-r}
\right).
$$

Este punto solo pertenece al cuadrado unidad si

$$
0\leq -\frac{T}{M}\leq 1.
$$

En particular, cuando

$$
x_2(A')=1,
$$

el punto $A'$ coincide con

$$
A=(0,1).
$$

Por tanto, geométricamente, $A'$ entra o sale del cuadrado unidad a través del punto $A$.

---

### Estado $B'$

De forma análoga, el punto $B'$ aparece sobre el borde

$$
x_2=0.
$$

Por tanto, en $B'$ la población 2 es completamente defectiva. La población 1 puede tener una fracción intermedia de cooperadores. Escribimos

$$
B'=(x_1(B'),0).
$$

Este punto se obtiene como el cruce entre la nulclina interior de $F_1=0$ y la nulclina

$$
x_2=0.
$$

De la primera parte del apéndice teníamos

$$
x_1=
-\frac{S+pNx_2}{\beta M},
$$

donde

$$
M=b-1-r,
\qquad
N=b-1-\varepsilon,
\qquad
S=\beta r+p\varepsilon.
$$

Sustituyendo $x_2=0$, queda

$$
x_1(B')
=
-\frac{S}{\beta M}.
$$

Por tanto,

$$
B'=
\left(
-\frac{S}{\beta M},
0
\right).
$$

Es decir,

$$
B'=
\left(
-\frac{\beta r+p\varepsilon}{\beta(b-1-r)},
0
\right).
$$

Este punto solo pertenece al cuadrado unidad si

$$
0\leq -\frac{S}{\beta M}\leq 1.
$$

En particular, cuando

$$
x_1(B')=1,
$$

el punto $B'$ coincide con

$$
B=(1,0).
$$

Por tanto, geométricamente, $B'$ entra o sale del cuadrado unidad a través del punto $B$.

---

### Relación con la pérdida de estabilidad de $A$ y $B$

En la sección anterior hemos visto que los estados polarizados pierden estabilidad en

$$
b_A^{\mathrm{up}}=1-\beta p\varepsilon,
$$

$$
b_B^{\mathrm{up}}=1-\frac{p\varepsilon}{\beta}.
$$

Estos valores también se pueden interpretar geométricamente:

- en $b=b_A^{\mathrm{up}}$, el punto $A'$ coincide con $A$;
- en $b=b_B^{\mathrm{up}}$, el punto $B'$ coincide con $B$.

Por eso el paper describe la pérdida de estabilidad de los estados polarizados como una transición hacia estados cuasipolarizados.

Sin embargo, todavía no hemos demostrado que $A'$ y $B'$ sean atractores. Para eso habrá que evaluar el jacobiano en estos puntos y estudiar sus autovalores. Además, dependiendo de los parámetros, puede ocurrir que alguno de estos puntos no pertenezca al cuadrado unidad o que no sea estable.

```python
# Comprobación de que A' coincide con A y B' coincide con B
# en los valores b_A^up y b_B^up

x2_Ap = -T/M
x1_Bp = -S/(beta*M)

A_prime = (0, x2_Ap)
B_prime = (x1_Bp, 0)

print("Coordenadas de A':")
display(A_prime)

print("Coordenadas de B':")
display(B_prime)

# A' coincide con A=(0,1) cuando b = b_A^up
check_zero(
    "A' coincide con A en b = b_A^up:",
    x2_Ap.subs(b, b_up_A) - 1
)

# B' coincide con B=(1,0) cuando b = b_B^up
check_zero(
    "B' coincide con B en b = b_B^up:",
    x1_Bp.subs(b, b_up_B) - 1
)
```

## 2.5. Pérdida de estabilidad de $A'$: valor $b_A^c$

Una vez obtenido el punto cuasipolarizado

$$
A'=
\left(
0,
-\frac{T}{M}
\right),
$$

estudiamos cuándo pierde estabilidad. En esta sección no clasificamos todavía todos los casos en los que $A'$ pertenece al cuadrado unidad; eso se tendrá en cuenta más adelante al ordenar los distintos escenarios.

Recordemos que hemos escrito

$$
F_1=x_1(1-x_1)g_1,
\qquad
F_2=x_2(1-x_2)g_2,
$$

donde, usando

$$
M=b-1-r,
\qquad
N=b-1-\varepsilon,
$$

$$
S=\beta r+p\varepsilon,
\qquad
T=r+\beta p\varepsilon,
$$

las funciones $g_1$ y $g_2$ pueden escribirse como

$$
g_1=-\beta Mx_1-pNx_2-S,
$$

$$
g_2=-\beta pNx_1-Mx_2-T.
$$

Por tanto, el jacobiano general es

$$
J(x_1,x_2)=
\begin{pmatrix}
(1-2x_1)g_1-\beta Mx_1(1-x_1)
&
-pNx_1(1-x_1)
\\[0.8em]
-\beta pNx_2(1-x_2)
&
(1-2x_2)g_2-Mx_2(1-x_2)
\end{pmatrix}.
$$

Ahora evaluamos en

$$
A'=
\left(
0,
-\frac{T}{M}
\right).
$$

Como $A'$ se obtiene imponiendo la nulclina interior de $F_2=0$, se cumple

$$
g_2(A')=0.
$$

Además, como $x_1(A')=0$, los términos proporcionales a $x_1(1-x_1)$ se anulan. Así, el jacobiano en $A'$ queda

$$
J(A')=
\begin{pmatrix}
g_1(A') & 0\\[0.6em]
-\beta pNx_2(A')[1-x_2(A')] & -Mx_2(A')[1-x_2(A')]
\end{pmatrix}.
$$

Es una matriz triangular. Por tanto, sus autovalores son las entradas diagonales:

$$
\lambda_{A',\perp}=g_1(A'),
$$

$$
\lambda_{A',\parallel}=-Mx_2(A')[1-x_2(A')].
$$

Calculemos primero $g_1(A')$. Como

$$
x_1(A')=0,
\qquad
x_2(A')=-\frac{T}{M},
$$

tenemos

$$
g_1(A')
=
-pN\left(-\frac{T}{M}\right)-S.
$$

Por tanto,

$$
g_1(A')
=
\frac{pNT}{M}-S
=
\frac{pNT-MS}{M}.
$$

Luego

$$
\lambda_{A',\perp}
=
\frac{pNT-MS}{M}.
$$

Este es el autovalor relevante para obtener $b_A^c$.

El otro autovalor,

$$
\lambda_{A',\parallel}
=
-Mx_2(A')[1-x_2(A')],
$$

describe perturbaciones a lo largo del borde $x_1=0$. Es decir, controla si el punto atrae o repele dentro del propio borde.

En cambio, la bifurcación $b_A^c$ corresponde a la pérdida de estabilidad de $A'$ frente al punto interior $E$. Esa pérdida ocurre en la dirección transversal al borde, es decir, en la dirección que permite pasar de $x_1=0$ a $x_1>0$.

Por eso, para calcular $b_A^c$ imponemos

$$
\lambda_{A',\perp}=0.
$$

Así,

$$
\frac{pNT-MS}{M}=0.
$$

Fuera del caso degenerado $M=0$, esto equivale a

$$
pNT-MS=0.
$$

O, multiplicando por $-1$,

$$
MS-pNT=0.
$$

Sustituyendo

$$
M=b-1-r,
\qquad
N=b-1-\varepsilon,
$$

$$
S=\beta r+p\varepsilon,
\qquad
T=r+\beta p\varepsilon,
$$

y despejando $b$, se obtiene

$$
b_A^c
=
1+
\frac{
\beta\left[r^2-(p\varepsilon)^2\right]
}
{
(\beta r+p\varepsilon)-p(r+\beta p\varepsilon)
}.
\qquad \text{(A19)}
$$

En resumen: $A'$ tiene dos autovalores, pero solo uno es relevante para esta bifurcación. El autovalor tangencial describe la estabilidad dentro del borde $x_1=0$, mientras que el autovalor transversal es el que se anula cuando $A'$ pierde estabilidad frente al equilibrio interior $E$.

```python
# Jacobiano en A' y autovalores relevantes

x2_Ap = -T/M
A_prime = {x1: 0, x2: x2_Ap}

J_Ap = simp(J_base.subs(A_prime))

print("Jacobiano evaluado en A':")
display(J_Ap)

# Como el jacobiano es triangular, los autovalores son las entradas diagonales
lambda_Ap_perp = simp(J_Ap[0, 0])
lambda_Ap_parallel = simp(J_Ap[1, 1])

print("Autovalor transversal:")
display(sp.Eq(sp.Symbol("lambda_Ap_perp"), lambda_Ap_perp))

print("Autovalor tangencial:")
display(sp.Eq(sp.Symbol("lambda_Ap_parallel"), lambda_Ap_parallel))

# Condición de bifurcación: autovalor transversal nulo
b_c_A = 1 + beta*(r**2 - (p*eps)**2)/((beta*r + p*eps) - p*(r + beta*p*eps))

b_c_A_obtenido = sp.solve(sp.Eq(M*S - p*N*T, 0), b)[0]

print("Valor crítico b_A^c:")
display(sp.Eq(sp.Symbol("b_A^c"), simp(b_c_A_obtenido)))

check_zero(
    "Comprobación de b_A^c:",
    b_c_A_obtenido - b_c_A
)

check_zero(
    "Condición MS-pNT evaluada en b_A^c:",
    (M*S - p*N*T).subs(b, b_c_A)
)
```

## 2.6. Pérdida de estabilidad de $B'$: valor $b_B^c$

De forma análoga, partimos del punto cuasipolarizado

$$
B'=
\left(
-\frac{S}{\beta M},
0
\right).
$$

En esta sección no clasificamos todavía todos los casos en los que $B'$ pertenece al cuadrado unidad. Supondremos que estamos en un escenario donde $B'$ es un punto fijo físico, y estudiaremos cuándo pierde estabilidad.

Recordemos que

$$
F_1=x_1(1-x_1)g_1,
\qquad
F_2=x_2(1-x_2)g_2,
$$

con

$$
g_1=-\beta Mx_1-pNx_2-S,
$$

$$
g_2=-\beta pNx_1-Mx_2-T,
$$

donde

$$
M=b-1-r,
\qquad
N=b-1-\varepsilon,
$$

$$
S=\beta r+p\varepsilon,
\qquad
T=r+\beta p\varepsilon.
$$

El jacobiano general es

$$
J(x_1,x_2)=
\begin{pmatrix}
(1-2x_1)g_1-\beta Mx_1(1-x_1)
&
-pNx_1(1-x_1)
\\[0.8em]
-\beta pNx_2(1-x_2)
&
(1-2x_2)g_2-Mx_2(1-x_2)
\end{pmatrix}.
$$

Ahora evaluamos en

$$
B'=
\left(
-\frac{S}{\beta M},
0
\right).
$$

Como $B'$ se obtiene imponiendo la nulclina interior de $F_1=0$, se cumple

$$
g_1(B')=0.
$$

Además, como

$$
x_2(B')=0,
$$

los términos proporcionales a $x_2(1-x_2)$ se anulan. Por tanto, el jacobiano en $B'$ queda

$$
J(B')=
\begin{pmatrix}
-\beta Mx_1(B')[1-x_1(B')] & -pNx_1(B')[1-x_1(B')]
\\[0.8em]
0 & g_2(B')
\end{pmatrix}.
$$

De nuevo es una matriz triangular. Por tanto, sus autovalores son las entradas diagonales:

$$
\lambda_{B',\parallel}
=
-\beta Mx_1(B')[1-x_1(B')],
$$

$$
\lambda_{B',\perp}
=
g_2(B').
$$

La distinción entre ambos autovalores es geométrica. El punto $B'$ está sobre el borde horizontal

$$
x_2=0.
$$

Una perturbación con

$$
\delta x_2=0
$$

permanece sobre ese borde, y por tanto es tangencial. En cambio, una perturbación con

$$
\delta x_2\neq 0
$$

saca al sistema del borde y lo introduce en el interior del cuadrado.

Por eso

$$
\lambda_{B',\parallel}
=
-\beta Mx_1(B')[1-x_1(B')]
$$

controla la estabilidad a lo largo del borde $x_2=0$, mientras que

$$
\lambda_{B',\perp}
=
g_2(B')
$$

controla la estabilidad transversal al borde.

Ahora calculamos explícitamente $g_2(B')$. Como

$$
x_1(B')=-\frac{S}{\beta M},
\qquad
x_2(B')=0,
$$

tenemos

$$
g_2(B')
=
-\beta pN\left(-\frac{S}{\beta M}\right)-T.
$$

Por tanto,

$$
g_2(B')
=
\frac{pNS}{M}-T.
$$

Es decir,

$$
g_2(B')
=
\frac{pNS-MT}{M}.
$$

Luego

$$
\lambda_{B',\perp}
=
\frac{pNS-MT}{M}.
$$

Este es el autovalor relevante para la bifurcación $B'\to E$, porque dicha bifurcación corresponde a que un punto de borde pierda estabilidad frente al punto interior $E$.

Por tanto, imponemos

$$
\lambda_{B',\perp}=0.
$$

Así,

$$
\frac{pNS-MT}{M}=0.
$$

Fuera del caso degenerado $M=0$, esto equivale a

$$
pNS-MT=0.
$$

O, multiplicando por $-1$,

$$
MT-pNS=0.
$$

Sustituyendo

$$
M=b-1-r,
\qquad
N=b-1-\varepsilon,
$$

$$
S=\beta r+p\varepsilon,
\qquad
T=r+\beta p\varepsilon,
$$

y despejando $b$, se obtiene

$$
b_B^c
=
1+
\frac{
r^2-(p\varepsilon)^2
}
{
(r+\beta p\varepsilon)-p(\beta r+p\varepsilon)
}.
\qquad \text{(A18)}
$$

En resumen: $B'$ tiene dos autovalores. El tangencial controla la estabilidad dentro del borde $x_2=0$, mientras que el transversal es el que se anula cuando $B'$ pierde estabilidad frente al equilibrio interior $E$. Por eso, para obtener $b_B^c$, imponemos que se anule el autovalor transversal.

```python
# Jacobiano en B' y autovalores relevantes

x1_Bp = -S/(beta*M)
B_prime = {x1: x1_Bp, x2: 0}

J_Bp = simp(J_base.subs(B_prime))

print("Jacobiano evaluado en B':")
display(J_Bp)

# Como el jacobiano es triangular, los autovalores son las entradas diagonales
lambda_Bp_parallel = simp(J_Bp[0, 0])
lambda_Bp_perp = simp(J_Bp[1, 1])

print("Autovalor tangencial:")
display(sp.Eq(sp.Symbol("lambda_Bp_parallel"), lambda_Bp_parallel))

print("Autovalor transversal:")
display(sp.Eq(sp.Symbol("lambda_Bp_perp"), lambda_Bp_perp))

# Condición de bifurcación: autovalor transversal nulo
b_c_B = 1 + (r**2 - (p*eps)**2)/((r + beta*p*eps) - p*(beta*r + p*eps))

b_c_B_obtenido = sp.solve(sp.Eq(M*T - p*N*S, 0), b)[0]

print("Valor crítico b_B^c:")
display(sp.Eq(sp.Symbol("b_B^c"), simp(b_c_B_obtenido)))

check_zero(
    "Comprobación de b_B^c:",
    b_c_B_obtenido - b_c_B
)

check_zero(
    "Condición MT-pNS evaluada en b_B^c:",
    (M*T - p*N*S).subs(b, b_c_B)
)
```

## 2.7. Resumen de los valores críticos en $r$ y en $b$

Hasta ahora hemos obtenido los valores críticos que aparecen al estudiar la estabilidad de los estados $D$, $A$, $B$, $A'$ y $B'$.

Para el estado completamente defectivo $D$, los cambios de estabilidad vienen dados por

$$
r_c^A=-\beta p\varepsilon,
$$

$$
r_c^B=-\frac{p\varepsilon}{\beta}.
$$

Estos valores separan los casos en los que $D$ es estable, silla o fuente.

Para los estados polarizados $A$ y $B$, los valores en los que pierden estabilidad son

$$
b_A^{\mathrm{up}}=1-\beta p\varepsilon,
$$

$$
b_B^{\mathrm{up}}=1-\frac{p\varepsilon}{\beta}.
$$

Como

$$
\varepsilon<0,
\qquad
\beta>1,
$$

se cumple

$$
b_B^{\mathrm{up}}<b_A^{\mathrm{up}}.
$$

Por tanto, al aumentar $b$ desde $1^+$, el estado $B$ pierde estabilidad antes que el estado $A$.

Finalmente, para los estados cuasipolarizados $A'$ y $B'$, hemos obtenido

$$
b_A^c
=
1+
\frac{
\beta\left[r^2-(p\varepsilon)^2\right]
}
{
(\beta r+p\varepsilon)-p(r+\beta p\varepsilon)
},
$$

$$
b_B^c
=
1+
\frac{
r^2-(p\varepsilon)^2
}
{
(r+\beta p\varepsilon)-p(\beta r+p\varepsilon)
}.
$$

Estos últimos valores indican dónde perderían estabilidad $A'$ y $B'$ si dichas bifurcaciones existen en el escenario considerado. La clasificación completa depende todavía de si $A'$ y $B'$ pertenecen al cuadrado unidad y del orden relativo entre los distintos valores críticos.

```python
# Tabla simbólica de valores críticos

valores_criticos = sp.Matrix([
    [sp.Symbol("r_c^A"), r_c_A],
    [sp.Symbol("r_c^B"), r_c_B],
    [sp.Symbol("b_A^up"), b_up_A],
    [sp.Symbol("b_B^up"), b_up_B],
    [sp.Symbol("b_A^c"), b_c_A],
    [sp.Symbol("b_B^c"), b_c_B],
])

display(valores_criticos)
```

## 2.8. Valor crítico $\beta_c^A$

Además de variar $b$ y $r$, el paper también estudia cómo cambia el retrato de fases al variar la razón de tamaños

$$
\beta=\frac{N_1}{N_2}.
$$

El primer valor crítico en $\beta$ aparece al analizar el valor candidato

$$
b_A^c.
$$

Recordemos que este valor se obtuvo al imponer la anulación del autovalor relevante de $A'$:

$$
b_A^c
=
1+
\frac{
\beta\left[r^2-(p\varepsilon)^2\right]
}
{
(\beta r+p\varepsilon)-p(r+\beta p\varepsilon)
}.
$$

Por ahora no afirmamos todavía que esta bifurcación ocurra siempre dentro del cuadrado unidad. Lo que estudiamos aquí es cuándo el valor candidato $b_A^c$ es finito y cuándo se va formalmente a infinito.

Para ello analizamos el denominador de la fracción:

$$
D_A=(\beta r+p\varepsilon)-p(r+\beta p\varepsilon).
$$

Desarrollando,

$$
D_A=\beta r+p\varepsilon-pr-\beta p^2\varepsilon.
$$

Agrupando los términos proporcionales a $\beta$,

$$
D_A=\beta(r-p^2\varepsilon)-p(r-\varepsilon).
$$

El valor crítico $\beta_c^A$ se obtiene imponiendo

$$
D_A=0.
$$

Por tanto,

$$
\beta(r-p^2\varepsilon)=p(r-\varepsilon).
$$

Despejando,

$$
\beta_c^A
=
\frac{p(r-\varepsilon)}{r-p^2\varepsilon}.
\qquad \text{(A21)}
$$

La interpretación de este valor es la siguiente: cuando

$$
\beta=\beta_c^A,
$$

el denominador de $b_A^c$ se anula y, por tanto, el valor candidato $b_A^c$ se va formalmente a infinito.

Esto no significa todavía que hayamos demostrado una bifurcación concreta dentro del cuadrado unidad. Solo indica que cambia el comportamiento del valor crítico asociado a $A'$: a un lado de $\beta_c^A$, $b_A^c$ aparece como un valor finito; al otro lado, desaparece hacia

$$
b=\infty.
$$

Más adelante, al clasificar los escenarios, se decidirá cuándo este valor crítico corresponde realmente a una bifurcación observable en el cuadrado unidad.

```python
# Deducción de beta_c^A

D_A = (beta*r + p*eps) - p*(r + beta*p*eps)

beta_c_A = p*(r - eps)/(r - p**2*eps)

display(sp.Eq(sp.Symbol("D_A"), sp.factor(D_A)))
display(sp.Eq(sp.Symbol("beta_c^A"), beta_c_A))

check_zero(
    "D_A se anula en beta = beta_c^A",
    D_A.subs(beta, beta_c_A)
)
```

## 2.9. Valor crítico $\beta_c^B$

El segundo valor crítico en la razón de tamaños es

$$
\beta_c^B.
$$

Este valor aparece al comparar dos valores críticos que ya hemos obtenido:

$$
b_B^c,
\qquad
b_A^{\mathrm{up}}.
$$

Recordemos que

$$
b_B^c
=
1+
\frac{
r^2-(p\varepsilon)^2
}
{
(r+\beta p\varepsilon)-p(\beta r+p\varepsilon)
},
$$

mientras que

$$
b_A^{\mathrm{up}}
=
1-\beta p\varepsilon.
$$

El valor $\beta_c^B$ se obtiene imponiendo que estos dos valores coincidan:

$$
b_B^c=b_A^{\mathrm{up}}.
$$

Por tanto,

$$
1+
\frac{
r^2-(p\varepsilon)^2
}
{
(r+\beta p\varepsilon)-p(\beta r+p\varepsilon)
}
=
1-\beta p\varepsilon.
$$

Restando $1$ a ambos lados,

$$
\frac{
r^2-(p\varepsilon)^2
}
{
(r+\beta p\varepsilon)-p(\beta r+p\varepsilon)
}
=
-\beta p\varepsilon.
$$

Para simplificar la notación, definimos el denominador de $b_B^c$ como

$$
D_B=(r+\beta p\varepsilon)-p(\beta r+p\varepsilon).
$$

Entonces la ecuación anterior queda

$$
\frac{
r^2-(p\varepsilon)^2
}
{D_B}
=
-\beta p\varepsilon.
$$

Multiplicando por $D_B$,

$$
r^2-(p\varepsilon)^2
=
-\beta p\varepsilon D_B.
$$

Pasando todo al mismo lado,

$$
r^2-(p\varepsilon)^2+\beta p\varepsilon D_B=0.
$$

Ahora desarrollamos $D_B$:

$$
D_B=(r+\beta p\varepsilon)-p(\beta r+p\varepsilon).
$$

Por tanto,

$$
D_B=r+\beta p\varepsilon-\beta pr-p^2\varepsilon.
$$

Agrupando los términos con $\beta$,

$$
D_B=r-p^2\varepsilon+\beta p(\varepsilon-r).
$$

Equivalente:

$$
D_B=r-p^2\varepsilon-\beta p(r-\varepsilon).
$$

Sustituimos esta expresión en la ecuación

$$
r^2-(p\varepsilon)^2+\beta p\varepsilon D_B=0.
$$

Así,

$$
r^2-p^2\varepsilon^2
+
\beta p\varepsilon
\left[
r-p^2\varepsilon-\beta p(r-\varepsilon)
\right]
=0.
$$

Desarrollamos:

$$
r^2-p^2\varepsilon^2
+
\beta p\varepsilon(r-p^2\varepsilon)
-
\beta^2p^2\varepsilon(r-\varepsilon)
=0.
$$

Ordenando como ecuación cuadrática en $\beta$,

$$
-\beta^2p^2\varepsilon(r-\varepsilon)
+
\beta p\varepsilon(r-p^2\varepsilon)
+
\left(r^2-p^2\varepsilon^2\right)
=0.
$$

Multiplicamos por $-1$ para dejar el coeficiente cuadrático con el signo que usa el paper:

$$
\beta^2p^2\varepsilon(r-\varepsilon)
-
\beta p\varepsilon(r-p^2\varepsilon)
-
\left(r^2-p^2\varepsilon^2\right)
=0.
$$

El término lineal puede escribirse como

$$
-\beta p\varepsilon(r-p^2\varepsilon)
=
\beta p\varepsilon(p^2\varepsilon-r),
$$

y el término independiente como

$$
-\left(r^2-p^2\varepsilon^2\right)
=
p^2\varepsilon^2-r^2.
$$

Por tanto, la ecuación cuadrática queda

$$
p^2\varepsilon(r-\varepsilon)\beta^2
+
p\varepsilon(p^2\varepsilon-r)\beta
+
(p^2\varepsilon^2-r^2)
=0.
$$

Esta ecuación tiene la forma

$$
a\beta^2+c\beta+d=0,
$$

con

$$
a=p^2\varepsilon(r-\varepsilon),
$$

$$
c=p\varepsilon(p^2\varepsilon-r),
$$

$$
d=p^2\varepsilon^2-r^2.
$$

Aplicando la fórmula general de una ecuación cuadrática,

$$
\beta=
\frac{-c\pm\sqrt{c^2-4ad}}{2a}.
$$

El discriminante es

$$
\Delta_B=c^2-4ad.
$$

Sustituyendo $a$, $c$ y $d$,

$$
\Delta_B
=
\left[p\varepsilon(p^2\varepsilon-r)\right]^2
-
4
\left[p^2\varepsilon(r-\varepsilon)\right]
\left[p^2\varepsilon^2-r^2\right].
$$

Es decir,

$$
\Delta_B
=
p^2\varepsilon^2(p^2\varepsilon-r)^2
-
4p^2\varepsilon(r-\varepsilon)
\left(p^2\varepsilon^2-r^2\right).
$$

Por tanto,

$$
\beta=
\frac{
-p\varepsilon(p^2\varepsilon-r)
\pm
\sqrt{\Delta_B}
}
{
2p^2\varepsilon(r-\varepsilon)
}.
$$

De las dos raíces algebraicas, el paper toma la raíz con el signo negativo delante de la raíz cuadrada:

$$
\beta_c^B
=
\frac{
-p\varepsilon(p^2\varepsilon-r)-\sqrt{\Delta_B}
}
{
2p^2\varepsilon(r-\varepsilon)
}.
\qquad \text{(A22)}
$$

La razón es que $\beta$ no es un parámetro cualquiera, sino una razón de tamaños poblacionales:

$$
\beta=\frac{N_1}{N_2}>1.
$$

Por tanto, de las dos soluciones de la ecuación cuadrática solo es relevante la que pertenece al rango físico $\beta>1$.

La otra raíz queda fuera de ese rango en el intervalo de parámetros en el que se usa este valor crítico, por lo que no corresponde a una razón de tamaños admisible para la clasificación del retrato de fases.

Así, $\beta_c^B$ no representa una nueva bifurcación en $b$. Representa el valor de $\beta$ para el cual coinciden los dos valores críticos candidatos

$$
b_B^c
\qquad \text{y} \qquad
b_A^{\mathrm{up}}.
$$

Al cruzar este valor de $\beta$, cambia el orden relativo entre ambos valores críticos. La interpretación dinámica completa se deja para la clasificación final de escenarios.

```python
# Comprobación simbólica de beta_c^B

# Denominador de b_B^c
D_B = (r + beta*p*eps) - p*(beta*r + p*eps)

print("Denominador D_B de b_B^c:")
display(sp.factor(D_B))

# Condición que define beta_c^B:
# b_B^c = b_A^up

num_beta_B = sp.together(b_c_B - b_up_A).as_numer_denom()[0]

print("Numerador de b_B^c - b_A^up:")
display(sp.factor(num_beta_B))

# Reescribimos la condición como una ecuación cuadrática en beta:
# a beta^2 + c beta + d = 0

a_beta = p**2*eps*(r - eps)
c_beta = p*eps*(p**2*eps - r)
d_beta = p**2*eps**2 - r**2

cuadratica_beta = a_beta*beta**2 + c_beta*beta + d_beta

print("Ecuación cuadrática en beta:")
display(sp.Eq(cuadratica_beta, 0))

# Comprobamos que es equivalente al numerador anterior
# Puede diferir por un signo global, que no cambia la ecuación = 0

check_zero(
    "Equivalencia con la condición b_B^c = b_A^up:",
    num_beta_B + cuadratica_beta
)

# Discriminante de la cuadrática
Delta_B = c_beta**2 - 4*a_beta*d_beta

print("Discriminante Delta_B:")
display(sp.factor(Delta_B))

# Raíces de la ecuación cuadrática
beta_plus = (-c_beta + sp.sqrt(Delta_B))/(2*a_beta)
beta_minus = (-c_beta - sp.sqrt(Delta_B))/(2*a_beta)

print("Raíz con signo +:")
display(sp.Eq(sp.Symbol("beta_+"), beta_plus))

print("Raíz con signo -:")
display(sp.Eq(sp.Symbol("beta_-"), beta_minus))

# La raíz que usa el paper es beta_minus
beta_c_B = beta_minus

print("Raíz relevante del paper:")
display(sp.Eq(sp.Symbol("beta_c^B"), beta_c_B))

# Comprobación final:
check_zero(
    "Comprobación: beta_c^B satisface b_B^c = b_A^up:",
    num_beta_B.subs(beta, beta_c_B)
)
```

```python
# Ejemplo numérico para ver qué raíz cae en el rango físico beta > 1

valores_ejemplo = {
    p: sp.Rational(3, 10),      # p = 0.3
    eps: sp.Rational(-2, 5),    # epsilon = -0.4
    r: sp.Rational(1, 20)       # r = 0.05
}

print("Valores usados:")
display(valores_ejemplo)

beta_plus_num = sp.N(beta_plus.subs(valores_ejemplo))
beta_minus_num = sp.N(beta_minus.subs(valores_ejemplo))

print("Raíz con signo +:")
display(beta_plus_num)

print("Raíz con signo -:")
display(beta_minus_num)

print("Rango físico: beta > 1")
```

## 2.10. Escenario (i): $r>r_c^A$

El primer escenario del caso general aparece cuando

$$
r>r_c^A,
$$

donde

$$
r_c^A=-\beta p\varepsilon.
$$

Recordemos que $\varepsilon<0$, por lo que

$$
r_c^A>0.
$$

Este caso corresponde a valores suficientemente grandes de $r$. Físicamente, $r$ mide el pago asociado a las interacciones defectivas dentro de cada población. Por tanto, si $r$ es grande, la deserción dentro de cada población está suficientemente favorecida como para que el estado completamente defectivo sea estable.

El estado completamente defectivo es

$$
D=(0,0).
$$

Como $x_1$ y $x_2$ son fracciones de cooperadores, esto significa que ambas poblaciones son completamente defectivas:

$$
x_1=0,
\qquad
x_2=0.
$$

En la sección anterior vimos que los autovalores de $D$ son

$$
\lambda_1(D)=-(\beta r+p\varepsilon),
$$

$$
\lambda_2(D)=-(r+\beta p\varepsilon).
$$

Para que $D$ sea estable, ambos autovalores deben ser negativos. La condición más restrictiva es

$$
r>r_c^A=-\beta p\varepsilon.
$$

Por tanto, en este escenario $D$ es estable para todo $b>1$.

Ahora bien, para valores bajos de la tentación $b$, los estados polarizados

$$
A=(0,1),
\qquad
B=(1,0)
$$

también son estables. Por eso, al principio del escenario coexisten tres atractores:

$$
D,A,B.
$$

Esto quiere decir que el estado final del sistema depende de la condición inicial: unas condiciones iniciales pueden acabar en $D$, otras en $A$ y otras en $B$.

Al aumentar $b$, los estados polarizados pierden estabilidad. Como en el caso general $\beta>1$, los valores críticos no coinciden:

$$
b_B^{\mathrm{up}}
=
1-\frac{p\varepsilon}{\beta},
$$

$$
b_A^{\mathrm{up}}
=
1-\beta p\varepsilon.
$$

Además, como

$$
\varepsilon<0,
\qquad
\beta>1,
$$

se cumple

$$
b_B^{\mathrm{up}}<b_A^{\mathrm{up}}.
$$

Por tanto, al aumentar $b$ desde $1^+$, el primer estado polarizado que pierde estabilidad es $B$.

En

$$
b=b_B^{\mathrm{up}},
$$

el estado $B$ deja de ser atractor. A partir de ese valor, los atractores que quedan son

$$
D,A.
$$

Después, al llegar a

$$
b=b_A^{\mathrm{up}},
$$

también pierde estabilidad el estado $A$. Como $D$ era estable desde el principio y sigue siéndolo para todo $b>1$, queda finalmente como único atractor global.

La secuencia de atractores es

$$
D,A,B
\xrightarrow{\ b_B^{\mathrm{up}}\ }
D,A
\xrightarrow{\ b_A^{\mathrm{up}}\ }
D.
\qquad \text{(A20)}
$$

La interpretación es sencilla: en este régimen el estado completamente defectivo $D$ es suficientemente robusto. Los estados polarizados pueden existir como atractores para valores bajos de $b$, pero desaparecen al aumentar la tentación. Primero desaparece $B$, porque es menos robusto, y después desaparece $A$. Finalmente, toda la dinámica acaba dominada por la deserción completa.

Este escenario corresponde al caso (a) de la tabla del paper.

## 2.11. Escenarios (ii): $r_c^B<r<r_c^A$

Consideramos ahora el intervalo

$$
r_c^B<r<r_c^A.
$$

Recordemos que

$$
r_c^B=-\frac{p\varepsilon}{\beta},
\qquad
r_c^A=-\beta p\varepsilon.
$$

Como $\varepsilon<0$ y $\beta>1$, se cumple

$$
r_c^B<r_c^A.
$$

En este intervalo el estado completamente defectivo

$$
D=(0,0)
$$

es una silla. En efecto, sus autovalores son

$$
\lambda_1(D)=-(\beta r+p\varepsilon),
$$

$$
\lambda_2(D)=-(r+\beta p\varepsilon).
$$

Si

$$
r>r_c^B=-\frac{p\varepsilon}{\beta},
$$

entonces

$$
\beta r+p\varepsilon>0,
$$

y por tanto

$$
\lambda_1(D)<0.
$$

Pero si

$$
r<r_c^A=-\beta p\varepsilon,
$$

entonces

$$
r+\beta p\varepsilon<0,
$$

y por tanto

$$
\lambda_2(D)>0.
$$

Así, $D$ tiene un autovalor negativo y otro positivo: no es atractor, sino silla.

---

### ¿Por qué no aparece $B'$ como atractor?

En este intervalo se cumple

$$
r_c^B<r<r_c^A.
$$

Esto implica, en particular,

$$
S=\beta r+p\varepsilon>0,
$$

$$
T=r+\beta p\varepsilon<0.
$$

Recordemos que

$$
B'=
\left(
-\frac{S}{\beta M},
0
\right),
$$

donde

$$
M=b-1-r.
$$

Para que $B'$ pertenezca al cuadrado unidad necesitamos

$$
0\leq -\frac{S}{\beta M}\leq 1.
$$

Como $S>0$, la condición

$$
-\frac{S}{\beta M}\geq 0
$$

obliga a que

$$
M<0.
$$

Es decir,

$$
b-1-r<0.
$$

Además, la condición

$$
-\frac{S}{\beta M}\leq 1
$$

lleva a

$$
b\leq b_B^{\mathrm{up}}.
$$

Por tanto, $B'$ solo puede estar dentro del cuadrado antes de la bifurcación de $B$.

Pero en ese intervalo $B'$ no es estable. De hecho, el autovalor de $B'$ asociado al movimiento sobre el borde $x_2=0$ es

$$
\lambda_1(B')
=
-\beta Mx_1(B')\left[1-x_1(B')\right].
$$

Si $B'$ está dentro del cuadrado, entonces

$$
0<x_1(B')<1,
$$

y acabamos de ver que en ese intervalo

$$
M<0.
$$

Por tanto,

$$
-\beta Mx_1(B')\left[1-x_1(B')\right]>0.
$$

Así, $B'$ tiene un autovalor positivo y no puede ser atractor.

Esta es la razón por la que en los escenarios (ii) no aparece una rama atractora $B'$. El estado $B$ pierde estabilidad en

$$
b=b_B^{\mathrm{up}},
$$

pero no deja detrás un $B'$ estable. Por eso, tras esa bifurcación, el atractor que queda es $A$.

---
### ¿Qué ocurre con $A'$?

El punto

$$
A'=
\left(
0,
-\frac{T}{M}
\right)
$$

puede entrar en el cuadrado unidad después de que $A$ pierda estabilidad.

Como en este intervalo

$$
r_c^B<r<r_c^A,
$$

se cumple

$$
T=r+\beta p\varepsilon<0.
$$

Por tanto,

$$
-T>0.
$$

El punto $A'$ entra por el vértice $A=(0,1)$ cuando

$$
x_2(A')=1.
$$

Es decir,

$$
-\frac{T}{M}=1.
$$

Esta condición da precisamente

$$
b=b_A^{\mathrm{up}}.
$$

Para

$$
b>b_A^{\mathrm{up}},
$$

el denominador

$$
M=b-1-r
$$

aumenta, mientras que $T$ permanece fijo. Por tanto,

$$
0<x_2(A')<1.
$$

Así, para $b>b_A^{\mathrm{up}}$, el punto $A'$ queda dentro del cuadrado unidad.

Ahora bien, esto solo demuestra que $A'$ es un punto fijo físicamente admisible. Para saber si es atractor hay que mirar los autovalores del jacobiano en $A'$.

De la sección anterior teníamos que los autovalores de $A'$ son

$$
\lambda_1(A')
=
-Mx_2(A')\left[1-x_2(A')\right],
$$

$$
\lambda_2(A')
=
\frac{pNT-MS}{M}.
$$

Una vez que $A'$ está dentro del cuadrado, se cumple

$$
0<x_2(A')<1.
$$

Además, para $b>b_A^{\mathrm{up}}$ se tiene

$$
M>0.
$$

Por tanto,

$$
\lambda_1(A')<0.
$$

El autovalor que decide si $A'$ sigue siendo estable o pierde estabilidad es entonces

$$
\lambda_2(A')
=
\frac{pNT-MS}{M}.
$$

Este autovalor se anula cuando

$$
pNT-MS=0,
$$

o equivalentemente,

$$
MS-pNT=0.
$$

Esa es precisamente la condición que define

$$
b_A^c.
$$

Por tanto, en el intervalo en el que $A'$ está dentro del cuadrado, puede actuar como atractor hasta que, si existe, se alcanza el valor crítico $b_A^c$.

Por eso, después de que $A$ pierda estabilidad en $b_A^{\mathrm{up}}$, el candidato natural a atractor es $A'$. Pero su estabilidad completa depende de si aparece o no el valor crítico $b_A^c$ en el escenario considerado.

---

### ¿Por qué a veces se acaba en $A'$ y a veces en $E$?

Aquí entra el punto interior

$$
E=(x_1(E),x_2(E)).
$$

De la primera parte del apéndice teníamos

$$
x_1(E)
=
\frac{MS-pNT}
{\beta\left[p^2N^2-M^2\right]}.
$$

El punto interior $E$ entra o sale del cuadrado a través del borde

$$
x_1=0
$$

cuando

$$
x_1(E)=0.
$$

Suponiendo que el denominador no se anula, esto ocurre cuando se anula el numerador:

$$
MS-pNT=0.
$$

Pero esta es exactamente la condición que nos dio

$$
b_A^c.
$$

Por tanto, $b_A^c$ es el valor candidato en el que el punto interior $E$ toca el borde $x_1=0$, es decir, el borde donde está $A'$.

Ahora bien, que este valor exista a un $b$ finito y en el orden correcto no ocurre siempre. Eso depende de $\beta$.

El valor

$$
\beta_c^A
=
\frac{p(r-\varepsilon)}{r-p^2\varepsilon}
$$

marca cuándo el valor candidato $b_A^c$ se va a infinito. Por eso hay dos posibilidades.

---

### Escenario (ii1): $\beta>\beta_c^A$

Si

$$
\beta>\beta_c^A,
$$

el valor candidato $b_A^c$ no aparece como un valor finito relevante al aumentar $b$.

La secuencia es entonces:

1. Para valores bajos de $b$, los atractores son

$$
A,
\qquad
B.
$$

2. En

$$
b=b_B^{\mathrm{up}},
$$

el estado $B$ pierde estabilidad. Como $B'$ no es atractor en este intervalo, queda solamente

$$
A.
$$

3. En

$$
b=b_A^{\mathrm{up}},
$$

el estado $A$ pierde estabilidad y el atractor pasa a ser

$$
A'.
$$

Como no aparece un $b_A^c$ finito relevante, la secuencia termina en $A'$:

$$
A,B
\xrightarrow{\ b_B^{\mathrm{up}}\ }
A
\xrightarrow{\ b_A^{\mathrm{up}}\ }
A'.
\qquad \text{(A23)}
$$

Este escenario corresponde al caso (c) de la tabla del paper.

---

### Escenario (ii2): $\beta<\beta_c^A$

Si

$$
\beta<\beta_c^A,
$$

el valor candidato $b_A^c$ sí aparece a un valor finito y con el orden adecuado en la secuencia.

La secuencia inicial es igual que antes:

$$
A,B
\xrightarrow{\ b_B^{\mathrm{up}}\ }
A
\xrightarrow{\ b_A^{\mathrm{up}}\ }
A'.
$$

Pero ahora, al llegar a

$$
b=b_A^c,
$$

el punto interior $E$ entra en el cuadrado a través del borde donde está $A'$. A partir de ahí, el atractor pasa a ser $E$.

Por tanto, la secuencia completa es

$$
A,B
\xrightarrow{\ b_B^{\mathrm{up}}\ }
A
\xrightarrow{\ b_A^{\mathrm{up}}\ }
A'
\xrightarrow{\ b_A^c\ }
E.
\qquad \text{(A24)}
$$

Este escenario corresponde al caso (b) de la tabla del paper.

---

En resumen, en el intervalo

$$
r_c^B<r<r_c^A,
$$

no aparece una fase atractora $B'$ porque, cuando $B'$ está dentro del cuadrado, tiene un autovalor positivo y es inestable.

La dinámica queda entonces organizada por la rama

$$
A\longrightarrow A'.
$$

La diferencia entre los dos subcasos es si el punto interior $E$ llega a entrar en el cuadrado a un valor finito de $b$:

- si $\beta>\beta_c^A$, no aparece un $b_A^c$ finito relevante y la secuencia termina en $A'$;
- si $\beta<\beta_c^A$, aparece $b_A^c$ y la secuencia termina en $E$.

## 2.12. Escenarios (iii): $r<r_c^B$

Finalmente consideramos el caso

$$
r<r_c^B.
$$

Recordemos que

$$
r_c^B=-\frac{p\varepsilon}{\beta},
\qquad
r_c^A=-\beta p\varepsilon,
$$

y que, como $\varepsilon<0$ y $\beta>1$,

$$
r_c^B<r_c^A.
$$

Por tanto, si

$$
r<r_c^B,
$$

entonces también se cumple

$$
r<r_c^A.
$$

En este intervalo, el estado completamente defectivo

$$
D=(0,0)
$$

es una fuente.

En efecto, los autovalores de $D$ son

$$
\lambda_1(D)=-(\beta r+p\varepsilon),
$$

$$
\lambda_2(D)=-(r+\beta p\varepsilon).
$$

Como

$$
r<r_c^B=-\frac{p\varepsilon}{\beta},
$$

se tiene

$$
\beta r+p\varepsilon<0,
$$

y por tanto

$$
\lambda_1(D)>0.
$$

Además, como también

$$
r<r_c^A=-\beta p\varepsilon,
$$

se tiene

$$
r+\beta p\varepsilon<0,
$$

y por tanto

$$
\lambda_2(D)>0.
$$

Así, los dos autovalores son positivos y $D$ es una fuente. Esto significa que la deserción completa no es atractora: pequeñas fluctuaciones cooperativas alejan al sistema de $D$.

---

### Entrada de $A'$ y $B'$ en el cuadrado

En este régimen se cumple

$$
S=\beta r+p\varepsilon<0,
$$

$$
T=r+\beta p\varepsilon<0.
$$

Recordemos que

$$
A'=
\left(
0,
-\frac{T}{M}
\right),
$$

$$
B'=
\left(
-\frac{S}{\beta M},
0
\right),
$$

donde

$$
M=b-1-r.
$$

Como $T<0$, se tiene

$$
-T>0.
$$

Por tanto, $A'$ puede entrar en el cuadrado unidad cuando

$$
0<-\frac{T}{M}<1.
$$

El punto $A'$ coincide con $A=(0,1)$ cuando

$$
-\frac{T}{M}=1.
$$

Esa condición da

$$
b=b_A^{\mathrm{up}}.
$$

Para

$$
b>b_A^{\mathrm{up}},
$$

se tiene

$$
0<x_2(A')<1,
$$

de modo que $A'$ queda dentro del cuadrado unidad.

Análogamente, como $S<0$, se tiene

$$
-S>0.
$$

Por tanto, $B'$ puede entrar en el cuadrado unidad cuando

$$
0<-\frac{S}{\beta M}<1.
$$

El punto $B'$ coincide con $B=(1,0)$ cuando

$$
-\frac{S}{\beta M}=1.
$$

Esa condición da

$$
b=b_B^{\mathrm{up}}.
$$

Para

$$
b>b_B^{\mathrm{up}},
$$

se tiene

$$
0<x_1(B')<1,
$$

de modo que $B'$ queda dentro del cuadrado unidad.

Como siempre se cumple

$$
b_B^{\mathrm{up}}<b_A^{\mathrm{up}},
$$

el primer estado cuasipolarizado que entra en el cuadrado es $B'$. Después entra $A'$.

---

### Estabilidad de $B'$

Una vez $B'$ está dentro del cuadrado, no basta con saber que existe: hay que mirar sus autovalores.

De la sección anterior teníamos que los autovalores de $B'$ son

$$
\lambda_1(B')
=
-\beta Mx_1(B')\left[1-x_1(B')\right],
$$

$$
\lambda_2(B')
=
\frac{pNS-MT}{M}.
$$

Cuando $B'$ está dentro del cuadrado,

$$
0<x_1(B')<1.
$$

Además, para

$$
b>b_B^{\mathrm{up}},
$$

se tiene

$$
M>0.
$$

Por tanto,

$$
\lambda_1(B')<0.
$$

El autovalor que decide cuándo $B'$ pierde estabilidad es entonces

$$
\lambda_2(B')
=
\frac{pNS-MT}{M}.
$$

Este autovalor se anula cuando

$$
pNS-MT=0,
$$

o equivalentemente,

$$
MT-pNS=0.
$$

Esa condición define

$$
b_B^c.
$$

Así, en este régimen $B'$ entra en el cuadrado en $b_B^{\mathrm{up}}$ y puede actuar como atractor hasta que, si se alcanza antes otra bifurcación relevante, pierde estabilidad en $b_B^c$.

---

### Estabilidad de $A'$ y papel de $E$

Análogamente, una vez $A'$ está dentro del cuadrado, sus autovalores son

$$
\lambda_1(A')
=
-Mx_2(A')\left[1-x_2(A')\right],
$$

$$
\lambda_2(A')
=
\frac{pNT-MS}{M}.
$$

Cuando $A'$ está dentro del cuadrado,

$$
0<x_2(A')<1,
$$

y para

$$
b>b_A^{\mathrm{up}}
$$

se tiene

$$
M>0.
$$

Por tanto,

$$
\lambda_1(A')<0.
$$

El autovalor que decide cuándo $A'$ pierde estabilidad es

$$
\lambda_2(A')
=
\frac{pNT-MS}{M}.
$$

Este autovalor se anula cuando

$$
pNT-MS=0,
$$

o equivalentemente,

$$
MS-pNT=0.
$$

Esa condición define

$$
b_A^c.
$$

Además, estas mismas condiciones aparecen al estudiar las coordenadas del punto interior $E$.

De la primera parte del apéndice,

$$
x_1(E)
=
\frac{MS-pNT}
{\beta\left[p^2N^2-M^2\right]}.
$$

Por tanto, suponiendo que el denominador no se anula,

$$
MS-pNT=0
$$

corresponde al valor en el que el punto interior $E$ toca el borde

$$
x_1=0.
$$

Ese es precisamente el borde donde está $A'$.

De forma análoga,

$$
x_2(E)
=
\frac{MT-pNS}
{p^2N^2-M^2}.
$$

Por tanto,

$$
MT-pNS=0
$$

corresponde al valor en el que $E$ toca el borde

$$
x_2=0.
$$

Ese es precisamente el borde donde está $B'$.

Esto, por sí solo, no demuestra todavía que $E$ sea estable. Solo nos dice cuándo el punto interior entra o sale del cuadrado unidad.

La estabilidad se decide combinando esta información con el análisis de autovalores de $A'$ y $B'$: en $b_A^c$ se anula el autovalor relevante de $A'$, y en $b_B^c$ se anula el autovalor relevante de $B'$. Por eso estos valores indican posibles intercambios de estabilidad entre un punto de borde y el punto interior $E$.

Más adelante, al ordenar los escenarios, se especifica en qué casos este intercambio conduce efectivamente a que $E$ sea el atractor final.

---

### ¿Por qué aparecen cuatro subescenarios?

En el régimen que estamos estudiando,

$$
r<r_c^B,
$$

el punto $D$ es una fuente y los dos estados cuasipolarizados $A'$ y $B'$ pueden entrar en el cuadrado unidad.

El primer hecho importante es que siempre se cumple

$$
b_B^{\mathrm{up}}<b_A^{\mathrm{up}}.
$$

En efecto,

$$
b_A^{\mathrm{up}}=1-\beta p\varepsilon,
\qquad
b_B^{\mathrm{up}}=1-\frac{p\varepsilon}{\beta}.
$$

Por tanto,

$$
b_A^{\mathrm{up}}-b_B^{\mathrm{up}}
=
\left(1-\beta p\varepsilon\right)
-
\left(1-\frac{p\varepsilon}{\beta}\right).
$$

Cancelando,

$$
b_A^{\mathrm{up}}-b_B^{\mathrm{up}}
=
-\beta p\varepsilon+\frac{p\varepsilon}{\beta}.
$$

Factorizando,

$$
b_A^{\mathrm{up}}-b_B^{\mathrm{up}}
=
p\varepsilon\left(\frac{1}{\beta}-\beta\right).
$$

Como

$$
\varepsilon<0,
\qquad
p\geq 0,
\qquad
\beta>1,
$$

tenemos

$$
p\varepsilon<0,
\qquad
\frac{1}{\beta}-\beta<0.
$$

Por tanto,

$$
b_A^{\mathrm{up}}-b_B^{\mathrm{up}}>0.
$$

Luego queda demostrado que

$$
b_B^{\mathrm{up}}<b_A^{\mathrm{up}}.
$$

Así, al aumentar $b$ desde $1^+$, el primer cambio ocurre siempre en

$$
b=b_B^{\mathrm{up}}.
$$

En ese valor, $B$ pierde estabilidad y el punto $B'$ entra en el cuadrado unidad.

La cuestión es qué ocurre después. Hay dos decisiones independientes que determinan la secuencia completa.

---

#### Primera decisión: ¿qué ocurre antes, $b_B^c$ o $b_A^{\mathrm{up}}$?

Una vez que $B'$ ha entrado en el cuadrado, puede perder estabilidad en

$$
b_B^c.
$$

Pero también sabemos que $A$ pierde estabilidad en

$$
b_A^{\mathrm{up}},
$$

momento en el que entra $A'$.

Por tanto, hay dos posibilidades:

$$
b_B^c<b_A^{\mathrm{up}},
$$

o bien

$$
b_A^{\mathrm{up}}<b_B^c.
$$

El valor que separa estas dos posibilidades es

$$
\beta_c^B.
$$

Esto es precisamente lo que significa la definición de $\beta_c^B$: se obtuvo imponiendo

$$
b_B^c=b_A^{\mathrm{up}}.
$$

Por tanto:

- si $\beta>\beta_c^B$, entonces

$$
b_B^c<b_A^{\mathrm{up}},
$$

y $B'$ pierde estabilidad antes de que entre $A'$;

- si $\beta<\beta_c^B$, entonces

$$
b_A^{\mathrm{up}}<b_B^c,
$$

y $A'$ entra en el cuadrado antes de que $B'$ pierda estabilidad.

---

#### Segunda decisión: ¿aparece o no $b_A^c$ como valor finito?

La segunda cuestión es si el valor

$$
b_A^c
$$

aparece como un valor finito relevante.

Esto lo controla

$$
\beta_c^A.
$$

Este valor se obtuvo al imponer que el denominador de $b_A^c$ se anule. Por tanto:

- si $\beta>\beta_c^A$, el valor candidato $b_A^c$ se desplaza formalmente hacia

$$
b=\infty,
$$

y no aparece como bifurcación finita en la secuencia;

- si $\beta<\beta_c^A$, el valor $b_A^c$ aparece a un valor finito de $b$ y debe incluirse en la secuencia.

---

#### Combinación de ambas decisiones

Por tanto, hay dos preguntas binarias:

1. ¿Está $b_B^c$ antes o después de $b_A^{\mathrm{up}}$?

2. ¿Aparece o no aparece $b_A^c$ a valor finito?

Al combinar estas dos posibilidades aparecen cuatro subescenarios:

| Condición sobre $\beta$ | Orden relevante | ¿Aparece $b_A^c$? | Escenario |
|---|---|---|---|
| $\max(\beta_c^A,\beta_c^B)<\beta$ | $b_B^c<b_A^{\mathrm{up}}$ | No | (iii1) |
| $\beta_c^B<\beta<\beta_c^A$ | $b_B^c<b_A^{\mathrm{up}}$ | Sí | (iii2) |
| $\beta_c^A<\beta<\beta_c^B$ | $b_A^{\mathrm{up}}<b_B^c$ | No | (iii3) |
| $\beta<\min(\beta_c^A,\beta_c^B)$ | $b_A^{\mathrm{up}}<b_B^c$ | Sí | (iii4) |

Esta es la razón por la que el caso $r<r_c^B$ se divide en cuatro escenarios distintos.

---

### Escenario (iii1)

Supongamos primero que

$$
\max(\beta_c^A,\beta_c^B)<\beta.
$$

Esta condición implica dos cosas:

1. Como

$$
\beta>\beta_c^B,
$$

se tiene

$$
b_B^c<b_A^{\mathrm{up}}.
$$

Es decir, $B'$ pierde estabilidad antes de que $A'$ entre en el cuadrado.

2. Como

$$
\beta>\beta_c^A,
$$

el valor $b_A^c$ no aparece como valor finito relevante.

La secuencia se entiende así:

Para valores bajos de $b$, los atractores son

$$
A,
\qquad
B.
$$

En

$$
b=b_B^{\mathrm{up}},
$$

el estado $B$ pierde estabilidad y aparece $B'$. Quedan como atractores

$$
A,
\qquad
B'.
$$

Después, como

$$
b_B^c<b_A^{\mathrm{up}},
$$

el estado $B'$ pierde estabilidad antes de que aparezca $A'$. Queda solo

$$
A.
$$

Finalmente, en

$$
b=b_A^{\mathrm{up}},
$$

el estado $A$ pierde estabilidad y aparece $A'$.

Como no hay un $b_A^c$ finito relevante, la secuencia termina en $A'$:

$$
A,B
\xrightarrow{\ b_B^{\mathrm{up}}\ }
A,B'
\xrightarrow{\ b_B^c\ }
A
\xrightarrow{\ b_A^{\mathrm{up}}\ }
A'.
\qquad \text{(A25)}
$$

Este escenario corresponde al caso (d) de la tabla.

---

### Escenario (iii2)

Ahora supongamos que

$$
\beta_c^B<\beta<\beta_c^A.
$$

Entonces:

1. Como

$$
\beta>\beta_c^B,
$$

se tiene

$$
b_B^c<b_A^{\mathrm{up}}.
$$

Por tanto, $B'$ pierde estabilidad antes de que $A'$ entre en el cuadrado.

2. Como

$$
\beta<\beta_c^A,
$$

aparece un valor finito

$$
b_A^c.
$$

La secuencia empieza igual que en el caso anterior:

$$
A,B
\xrightarrow{\ b_B^{\mathrm{up}}\ }
A,B'
\xrightarrow{\ b_B^c\ }
A
\xrightarrow{\ b_A^{\mathrm{up}}\ }
A'.
$$

Pero ahora sí aparece posteriormente el valor

$$
b_A^c.
$$

En ese valor, el punto interior $E$ entra en el cuadrado por el borde donde está $A'$, y el atractor pasa a ser $E$.

Por tanto,

$$
A,B
\xrightarrow{\ b_B^{\mathrm{up}}\ }
A,B'
\xrightarrow{\ b_B^c\ }
A
\xrightarrow{\ b_A^{\mathrm{up}}\ }
A'
\xrightarrow{\ b_A^c\ }
E.
\qquad \text{(A26)}
$$

Este escenario corresponde al caso (e) de la tabla.

---

### Escenario (iii3)

Ahora consideramos

$$
\beta_c^A<\beta<\beta_c^B.
$$

Entonces:

1. Como

$$
\beta<\beta_c^B,
$$

se tiene

$$
b_A^{\mathrm{up}}<b_B^c.
$$

Por tanto, $A'$ entra en el cuadrado antes de que $B'$ pierda estabilidad.

2. Como

$$
\beta>\beta_c^A,
$$

el valor $b_A^c$ no aparece como valor finito relevante.

La secuencia es entonces:

Primero,

$$
A,B
\xrightarrow{\ b_B^{\mathrm{up}}\ }
A,B'.
$$

Después, como

$$
b_A^{\mathrm{up}}<b_B^c,
$$

el estado $A$ pierde estabilidad y aparece $A'$ antes de que $B'$ se desestabilice. Por tanto, durante un intervalo coexisten

$$
A',
\qquad
B'.
$$

Finalmente, en

$$
b=b_B^c,
$$

el estado $B'$ pierde estabilidad. Como no aparece un $b_A^c$ finito relevante, queda $A'$ como atractor final.

La secuencia es

$$
A,B
\xrightarrow{\ b_B^{\mathrm{up}}\ }
A,B'
\xrightarrow{\ b_A^{\mathrm{up}}\ }
A',B'
\xrightarrow{\ b_B^c\ }
A'.
\qquad \text{(A27)}
$$

Este escenario corresponde al caso (f) de la tabla.

---

### Escenario (iii4)

Finalmente, supongamos que

$$
\beta<\min(\beta_c^A,\beta_c^B).
$$

Entonces:

1. Como

$$
\beta<\beta_c^B,
$$

se tiene

$$
b_A^{\mathrm{up}}<b_B^c.
$$

Por tanto, $A'$ entra en el cuadrado antes de que $B'$ pierda estabilidad.

2. Como

$$
\beta<\beta_c^A,
$$

aparece un valor finito

$$
b_A^c.
$$

La secuencia inicial es igual que en el caso anterior:

$$
A,B
\xrightarrow{\ b_B^{\mathrm{up}}\ }
A,B'
\xrightarrow{\ b_A^{\mathrm{up}}\ }
A',B'.
$$

Después, en

$$
b=b_B^c,
$$

el estado $B'$ pierde estabilidad, quedando

$$
A'.
$$

Finalmente, como existe $b_A^c$, el punto interior $E$ entra en el cuadrado por el borde donde está $A'$, y el atractor final pasa a ser

$$
E.
$$

La secuencia completa es

$$
A,B
\xrightarrow{\ b_B^{\mathrm{up}}\ }
A,B'
\xrightarrow{\ b_A^{\mathrm{up}}\ }
A',B'
\xrightarrow{\ b_B^c\ }
A'
\xrightarrow{\ b_A^c\ }
E.
\qquad \text{(A28)}
$$

Este escenario corresponde al caso (g) de la tabla.

---

En resumen, cuando

$$
r<r_c^B,
$$

los dos estados cuasipolarizados pueden aparecer dentro del cuadrado unidad.

El estado $B'$ entra primero, porque

$$
b_B^{\mathrm{up}}<b_A^{\mathrm{up}}.
$$

Luego hay dos preguntas que determinan la secuencia:

1. ¿Se desestabiliza $B'$ antes o después de que entre $A'$?
   
   Esto lo decide $\beta_c^B$.

2. ¿Existe un valor finito $b_A^c$?
   
   Esto lo decide $\beta_c^A$.

Combinando ambas respuestas se obtienen los cuatro escenarios (A25)-(A28).

## 2.13. Resumen final de escenarios

El análisis anterior permite resumir las posibles secuencias de atractores al aumentar la tentación desde

$$
b=1^+.
$$

Primero, en el caso simétrico

$$
N_1=N_2,
\qquad
\beta=1,
$$

solo aparecen dos escenarios:

| Caso | Condición | Secuencia de atractores |
|---|---|---|
| Simétrico (s1) | $r>r_c=-p\varepsilon$ | $D,A,B \xrightarrow{\,b^{\mathrm{up}}\,} D$ |
| Simétrico (s2) | $r<r_c=-p\varepsilon$ | $A,B \xrightarrow{\,b^{\mathrm{up}}\,} A',B' \xrightarrow{\,b^c\,} E$ |

En el caso general

$$
N_1\neq N_2,
\qquad
\beta=\frac{N_1}{N_2}>1,
$$

la simetría entre $A$ y $B$ se rompe. Esto da lugar a los siete escenarios recogidos en la tabla del paper:

| Escenario | Condición | Secuencia de atractores |
|---|---|---|
| (a) | $r>r_c^A$ | $D,A,B \xrightarrow{\,b_B^{\mathrm{up}}\,} D,A \xrightarrow{\,b_A^{\mathrm{up}}\,} D$ |
| (b) | $r_c^B<r<r_c^A,\ \beta<\beta_c^A$ | $A,B \xrightarrow{\,b_B^{\mathrm{up}}\,} A \xrightarrow{\,b_A^{\mathrm{up}}\,} A' \xrightarrow{\,b_A^c\,} E$ |
| (c) | $r_c^B<r<r_c^A,\ \beta>\beta_c^A$ | $A,B \xrightarrow{\,b_B^{\mathrm{up}}\,} A \xrightarrow{\,b_A^{\mathrm{up}}\,} A'$ |
| (d) | $r<r_c^B,\ \max(\beta_c^A,\beta_c^B)<\beta$ | $A,B \xrightarrow{\,b_B^{\mathrm{up}}\,} A,B' \xrightarrow{\,b_B^c\,} A \xrightarrow{\,b_A^{\mathrm{up}}\,} A'$ |
| (e) | $r<r_c^B,\ \beta_c^B<\beta<\beta_c^A$ | $A,B \xrightarrow{\,b_B^{\mathrm{up}}\,} A,B' \xrightarrow{\,b_B^c\,} A \xrightarrow{\,b_A^{\mathrm{up}}\,} A' \xrightarrow{\,b_A^c\,} E$ |
| (f) | $r<r_c^B,\ \beta_c^A<\beta<\beta_c^B$ | $A,B \xrightarrow{\,b_B^{\mathrm{up}}\,} A,B' \xrightarrow{\,b_A^{\mathrm{up}}\,} A',B' \xrightarrow{\,b_B^c\,} A'$ |
| (g) | $r<r_c^B,\ \beta<\min(\beta_c^A,\beta_c^B)$ | $A,B \xrightarrow{\,b_B^{\mathrm{up}}\,} A,B' \xrightarrow{\,b_A^{\mathrm{up}}\,} A',B' \xrightarrow{\,b_B^c\,} A' \xrightarrow{\,b_A^c\,} E$ |

Cada flecha representa una bifurcación al aumentar $b$, y el valor escrito encima indica el valor crítico en el que ocurre.
