Por otro lado, rara vez se da el caso de un monopolio del crimen. Lo habitual es que haya cierto ambiente de competitividad, a menudo violenta y territorial, entre varias mafias. No obstante, al desarrollarse la competición a través de la violencia, la misma está regida por la ley del más fuerte: desfavorece la existencia de mafias pequeñas (i.e., débiles), a la par que propicia la subsistencia y boyanza de las más grandes y poderosas. Como consecuencia, el modelo oligárquico más simple (el de 2 mafias antagónicas) es también el más razonable si consideramos los puntos anteriores. Evidentemente, en la realidad —donde la dimensión espacial juega un papel fundamental a la hora de limitar el área de influencia y el alcance de las disputas territoriales— se dan situaciones enormemente más complejas, especialmente en ciudades o países donde el crimen organizado ha permeado todas las capas de la sociedad, y se encuentra severamente extendido y asentado. Por ejemplo, en Ecuador, un país donde los grupos de delincuencia organizada han crecido en poder y números a un ritmo vertiginoso en el transcurso de la última década, nos encontramos con grandes ciudades transformadas en verdaderos epicentros del crimen, como es el caso de Guayaquil (que debido a su puerto ha sido convertido en el punto de salida del tráfico de drogas, y el punto de entrada del tráfico de armas). Debido a su importancia como centro logístico de las operaciones del narco, que constituyen la principal fuente de ingresos de toda la estructura criminal, todas las mafias nacionales y transnacionales de Ecuador ‒sin importar su tamaño‒ coexisten en la ciudad, cada uno dentro de su territorio, pero en constante enfrentamiento.
