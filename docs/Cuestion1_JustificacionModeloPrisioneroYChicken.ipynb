*Justificación de la dinámica intramafia ≡ Dilema del Prisionero*

En las estructuras sociales vinculadas al crimen organizado, comúnmente conocidas como mafias, las interacciones entre miembros tienen un carácter jerárquico. Empezaremos sin suponer una cadena de mando preestablecida (crimen débilmente organizado). Esto hará que, si se encuentran dos integrantes de la misma mafia, ambos quieran tomar el puesto de poder desde el cual dar órdenes al otro. 

No obstante, si ambos toman esta determinación de manera inflexible, destinarán gran parte de sus esfuerzos a tratar de establecer dicha dominancia, en lugar de centrarse en las actividades que lucran a la mafia, o en la lucha contra mafias rivales. Además, tal y como estableceremos a continuación en la matriz de premios y castigos, supondremos que los criminales son propensos a emplear la violencia, incluso contra miembros de su misma mafia, con tal de imponer su dominancia. Por otro lado, rara vez se da el caso de un monopolio del crimen. Lo habitual es que haya cierto ambiente de competitividad, a menudo violenta y territorial, entre varias mafias. No obstante, al desarrollarse la competición a través de la violencia, la misma está regida por la ley del más fuerte: desfavorece la existencia de mafias pequeñas (i.e., débiles), a la par que propicia la subsistencia y boyanza de las más grandes y poderosas. Evidentemente, en la realidad, factores como la dimensión espacial juegan un papel fundamental a la hora de limitar el área de influencia y el alcance de las disputas territoriales; y existen fenómenos —como las disidencias internas y escisiones de bandas [1], o las alianzas y fusiones— que dan lugar a situaciones enormemente más complejas, especialmente en ciudades o países donde el crimen organizado ha permeado todas las capas de la sociedad, y se encuentra severamente extendido y asentado. Por ejemplo, en Ecuador, un país donde los grupos de delincuencia organizada han crecido en poder y números a un ritmo vertiginoso en el transcurso de la última década, nos encontramos con grandes ciudades transformadas en verdaderos epicentros del crimen. Este es el caso de Guayaquil: debido a su estratégico puerto, ha sido convertida en el punto de salida del tráfico de drogas hacia Europa, y el punto de entrada del tráfico de armas desde Estados Unidos. A raíz de su importancia como centro logístico de las operaciones del narco, que constituyen la principal fuente de ingresos de toda la estructura criminal, todas las mafias nacionales y transnacionales de Ecuador ‒sin importar su tamaño‒ coexisten en la ciudad: cada una tiene sus bastiones y sus áreas de influencia, pero las fronteras entre territorios son siempre puntos de fricción, a veces con barrios enteros en disputa.




No obstante, incluso en panoramas tan complejos, es de esperar que ‒si nos centramos en un área lo suficientemente pequeña‒ la competición se reduzca a un número pequeño de bandas, idealmente dos.
